```python
# ============================================================
# US FUNDING & MONEY-MARKET MONITOR
# CELL 1 — DATA + SPREAD CONSTRUCTION + VALIDATION
# ============================================================

import pandas as pd
import numpy as np
import requests
from io import StringIO

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 180)

# ============================================================
# 1. FRED DOWNLOAD FUNCTION
# ============================================================

def fred(series_id, start="1990-01-01"):
    url = (
        "https://fred.stlouisfed.org/graph/fredgraph.csv"
        f"?id={series_id}&cosd={start}"
    )

    df = pd.read_csv(url)

    df.columns = ["Date", series_id]
    df["Date"] = pd.to_datetime(df["Date"])
    df[series_id] = pd.to_numeric(df[series_id], errors="coerce")

    return df.set_index("Date")[series_id]


# ============================================================
# 2. SERIES MAP
# ============================================================

SERIES = {

    # --------------------------------------------------------
    # SECURED / OVERNIGHT FUNDING
    # --------------------------------------------------------
    "SOFR": "SOFR",
    "TGCR": "RRPONTSYAWARD",   # validated below; removed if invalid
    "EFFR": "EFFR",

    # --------------------------------------------------------
    # FED ADMINISTERED RATES
    # --------------------------------------------------------
    "IORB": "IORB",
    "ON_RRP": "RRPONTSYAWARD",

    # --------------------------------------------------------
    # COMMERCIAL PAPER — AA FINANCIAL
    # --------------------------------------------------------
    "AA_FIN_30D": "RIFSPPFAAD30NB",
    "AA_FIN_60D": "RIFSPPFAAD60NB",
    "AA_FIN_90D": "RIFSPPFAAD90NB",

    # --------------------------------------------------------
    # COMMERCIAL PAPER — AA NONFINANCIAL
    # --------------------------------------------------------
    "AA_NF_30D": "RIFSPPNAAD30NB",
    "AA_NF_60D": "RIFSPPNAAD60NB",
    "AA_NF_90D": "RIFSPPNAAD90NB",

    # --------------------------------------------------------
    # COMMERCIAL PAPER — A2/P2 NONFINANCIAL
    # --------------------------------------------------------
    "A2P2_NF_30D": "RIFSPPNA2P2D30NB",
    "A2P2_NF_60D": "RIFSPPNA2P2D60NB",
    "A2P2_NF_90D": "RIFSPPNA2P2D90NB",
}

# IMPORTANT:
# TGCR/BGCR are better pulled from NY Fed directly.
# Therefore remove placeholder TGCR before download.
SERIES.pop("TGCR")


# ============================================================
# 3. DOWNLOAD FRED SERIES
# ============================================================

raw = {}

print("DOWNLOADING DATA...\n")

for name, fred_id in SERIES.items():

    try:
        s = fred(fred_id)
        raw[name] = s

        valid = s.dropna()

        print(
            f"{name:<15} "
            f"{valid.index.min().date()} -> "
            f"{valid.index.max().date()}   "
            f"N={len(valid):,}"
        )

    except Exception as e:

        print(f"{name:<15} FAILED: {e}")


# ============================================================
# 4. NY FED API — SOFR / TGCR / BGCR
# ============================================================

def nyfed_reference_rate(rate_type):

    url = (
        "https://markets.newyorkfed.org/api/rates/"
        f"secured/{rate_type}/search.json"
        "?startDate=2018-01-01"
        "&type=rate"
    )

    r = requests.get(url, timeout=30)
    r.raise_for_status()

    js = r.json()

    # NY Fed response structure
    records = js["refRates"]

    df = pd.DataFrame(records)

    # Standard fields are effectiveDate and percentRate
    df["Date"] = pd.to_datetime(df["effectiveDate"])
    df["Rate"] = pd.to_numeric(df["percentRate"], errors="coerce")

    return (
        df[["Date", "Rate"]]
        .dropna()
        .drop_duplicates("Date")
        .set_index("Date")["Rate"]
        .sort_index()
    )


for name in ["SOFR", "TGCR", "BGCR"]:

    try:

        s = nyfed_reference_rate(name.lower())

        # Replace FRED SOFR with direct NY Fed source
        raw[name] = s

        print(
            f"{name:<15} "
            f"{s.index.min().date()} -> "
            f"{s.index.max().date()}   "
            f"N={len(s):,}  [NY FED]"
        )

    except Exception as e:

        print(f"{name:<15} NY FED FAILED: {e}")


# ============================================================
# 5. COMBINE RAW DATA
# ============================================================

rates = pd.concat(raw, axis=1).sort_index()


# ============================================================
# 6. PAIRWISE SPREAD FUNCTION
#
# IMPORTANT:
# Every spread starts at ITS OWN earliest valid overlap.
# We do NOT truncate everything to one common sample.
#
# All rates are already expressed in percentage points.
# Multiply difference by 100 -> basis points.
# ============================================================

def make_spread(a, b):

    pair = rates[[a, b]].dropna()

    spread = (pair[a] - pair[b]) * 100

    return spread


spreads = {}


# ============================================================
# 7. OVERNIGHT / REPO BASIS
# ============================================================

spreads["SOFR-EFFR"] = make_spread("SOFR", "EFFR")

spreads["TGCR-SOFR"] = make_spread("TGCR", "SOFR")

spreads["BGCR-TGCR"] = make_spread("BGCR", "TGCR")


# ============================================================
# 8. POLICY / ADMINISTERED-RATE BASIS
# ============================================================

spreads["SOFR-IORB"] = make_spread("SOFR", "IORB")

spreads["EFFR-IORB"] = make_spread("EFFR", "IORB")

spreads["SOFR-ON_RRP"] = make_spread("SOFR", "ON_RRP")


# ============================================================
# 9. COMMERCIAL PAPER CREDIT / LIQUIDITY SPREADS
#
# A2/P2 NF minus AA NF
# ============================================================

for maturity in ["30D", "60D", "90D"]:

    spreads[f"A2P2-AA_NF_{maturity}"] = make_spread(
        f"A2P2_NF_{maturity}",
        f"AA_NF_{maturity}"
    )


# ============================================================
# 10. FINANCIAL vs NONFINANCIAL CP
# ============================================================

for maturity in ["30D", "60D", "90D"]:

    spreads[f"AA_FIN-AA_NF_{maturity}"] = make_spread(
        f"AA_FIN_{maturity}",
        f"AA_NF_{maturity}"
    )


# ============================================================
# 11. CP TERM SPREADS
#
# 90D minus 30D within each credit category
# ============================================================

spreads["AA_FIN_90D-30D"] = make_spread(
    "AA_FIN_90D",
    "AA_FIN_30D"
)

spreads["AA_NF_90D-30D"] = make_spread(
    "AA_NF_90D",
    "AA_NF_30D"
)

spreads["A2P2_NF_90D-30D"] = make_spread(
    "A2P2_NF_90D",
    "A2P2_NF_30D"
)


# ============================================================
# 12. VALIDATION TABLE — RAW SERIES
# ============================================================

raw_validation = []

for col in rates.columns:

    x = rates[col].dropna()

    raw_validation.append({

        "Series": col,
        "Start": x.index.min().date(),
        "End": x.index.max().date(),
        "Obs": len(x),
        "Latest": round(x.iloc[-1], 4),
        "Min": round(x.min(), 4),
        "Max": round(x.max(), 4)

    })

raw_validation = pd.DataFrame(raw_validation)


# ============================================================
# 13. VALIDATION TABLE — SPREADS
# ============================================================

spread_validation = []

for name, x in spreads.items():

    x = x.dropna()

    spread_validation.append({

        "Spread": name,
        "Start": x.index.min().date(),
        "End": x.index.max().date(),
        "Obs": len(x),
        "Latest_bp": round(x.iloc[-1], 2),
        "Mean_bp": round(x.mean(), 2),
        "Min_bp": round(x.min(), 2),
        "Max_bp": round(x.max(), 2)

    })

spread_validation = pd.DataFrame(spread_validation)


# ============================================================
# 14. OUTPUT
# ============================================================

print("\n\n================ RAW SERIES ================\n")
display(raw_validation)

print("\n\n================ CONSTRUCTED SPREADS ================\n")
display(spread_validation)

print("\n\n================ LATEST RAW RATES ================\n")
display(
    rates.ffill()
         .iloc[-1]
         .sort_values(ascending=False)
         .to_frame("Rate (%)")
)

print("\nDone.")
```

```python
# ============================================================
# US FUNDING & MONEY-MARKET MONITOR
# CELL 2 — PERSISTENCE / Z-SCORES / EXTREMES
#
# Requires Cell 1 to have already been run.
# ============================================================

!pip -q install statsmodels

import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 220)
pd.set_option("display.max_rows", 200)

# ============================================================
# 1. HELPERS
# ============================================================

def percentile_rank(x):
    """Historical percentile of latest observation."""
    x = x.dropna()
    if len(x) == 0:
        return np.nan

    latest = x.iloc[-1]
    return 100 * (x <= latest).mean()


def zscore_latest(x, window=None):
    """
    Latest z-score.
    window=None -> full history
    """
    x = x.dropna()

    if window is not None:
        x = x.iloc[-window:]

    if len(x) < 2:
        return np.nan

    sd = x.std(ddof=1)

    if sd == 0 or np.isnan(sd):
        return np.nan

    return (x.iloc[-1] - x.mean()) / sd


def ar1_stats(x):
    """
    AR(1):
        x_t = c + phi*x_(t-1) + error

    Returns phi and implied OU half-life in observations.
    """
    x = x.dropna()

    if len(x) < 30:
        return np.nan, np.nan

    y = x.iloc[1:].values
    lag = x.iloc[:-1].values

    X = sm.add_constant(lag)

    model = sm.OLS(y, X).fit()

    phi = model.params[1]

    # Valid mean-reverting AR(1) region
    if 0 < phi < 1:
        half_life = -np.log(2) / np.log(phi)
    else:
        half_life = np.nan

    return phi, half_life


def adf_pvalue(x):
    """
    Augmented Dickey-Fuller test.
    H0 = unit root / non-stationary.
    """
    x = x.dropna()

    if len(x) < 30:
        return np.nan

    try:
        result = adfuller(
            x.values,
            regression="c",
            autolag="AIC"
        )
        return result[1]

    except Exception:
        return np.nan


def rolling_vol_latest(x, window):
    """
    Standard deviation of spread levels over latest window.
    Units remain basis points.
    """
    x = x.dropna()

    if len(x) < window:
        return np.nan

    return x.iloc[-window:].std(ddof=1)


# ============================================================
# 2. MAIN ANALYTICS
# ============================================================

analytics = []

for name, s in spreads.items():

    x = s.dropna().sort_index()

    phi, half_life = ar1_stats(x)

    current = x.iloc[-1]

    full_mean = x.mean()
    full_median = x.median()
    full_sd = x.std(ddof=1)

    analytics.append({

        "Spread": name,

        "Start": x.index.min().date(),
        "End": x.index.max().date(),
        "Obs": len(x),

        "Current_bp": current,

        "Mean_bp": full_mean,
        "Median_bp": full_median,

        "Percentile": percentile_rank(x),

        "Z_Full": zscore_latest(x),

        # ~1 trading year
        "Z_1Y": zscore_latest(x, 252),

        # ~3 months
        "Z_63D": zscore_latest(x, 63),

        "AR1_phi": phi,

        "HalfLife_obs": half_life,

        "ADF_p": adf_pvalue(x),

        "Vol_21D_bp": rolling_vol_latest(x, 21),

        "Vol_63D_bp": rolling_vol_latest(x, 63),

        "Hist_Min_bp": x.min(),
        "Hist_Max_bp": x.max()

    })


analytics = pd.DataFrame(analytics)


# ============================================================
# 3. ROUND DISPLAY COPY
# ============================================================

analytics_display = analytics.copy()

round_cols = [
    "Current_bp",
    "Mean_bp",
    "Median_bp",
    "Percentile",
    "Z_Full",
    "Z_1Y",
    "Z_63D",
    "AR1_phi",
    "HalfLife_obs",
    "ADF_p",
    "Vol_21D_bp",
    "Vol_63D_bp",
    "Hist_Min_bp",
    "Hist_Max_bp"
]

analytics_display[round_cols] = (
    analytics_display[round_cols].round(3)
)


# ============================================================
# 4. EXTREME OBSERVATIONS
#
# Top 5 highest + lowest observations for every spread.
# This lets us inspect the strange CP negatives and repo spikes.
# ============================================================

extremes = []

for name, s in spreads.items():

    x = s.dropna().sort_index()

    highest = x.nlargest(5)
    lowest = x.nsmallest(5)

    for date, value in highest.items():

        extremes.append({
            "Spread": name,
            "Side": "HIGH",
            "Date": date.date(),
            "Value_bp": value
        })

    for date, value in lowest.items():

        extremes.append({
            "Spread": name,
            "Side": "LOW",
            "Date": date.date(),
            "Value_bp": value
        })


extremes = pd.DataFrame(extremes)

extremes["Value_bp"] = extremes["Value_bp"].round(2)


# ============================================================
# 5. STATIONARITY / PERSISTENCE FLAGS
#
# These are descriptive flags, NOT trading signals.
# ============================================================

def stationarity_label(p):

    if pd.isna(p):
        return "N/A"

    if p < 0.01:
        return "Strong"

    if p < 0.05:
        return "Yes"

    if p < 0.10:
        return "Weak"

    return "No"


def persistence_label(phi):

    if pd.isna(phi):
        return "N/A"

    if phi >= 0.95:
        return "Very High"

    if phi >= 0.80:
        return "High"

    if phi >= 0.50:
        return "Moderate"

    if phi > 0:
        return "Low"

    return "Negative"


analytics_display["ADF_Stationary"] = (
    analytics["ADF_p"].apply(stationarity_label)
)

analytics_display["Persistence"] = (
    analytics["AR1_phi"].apply(persistence_label)
)


# ============================================================
# 6. CURRENT EXTREMENESS RANKING
#
# Absolute full-history z-score.
# ============================================================

current_extremes = (
    analytics_display
    .assign(Abs_Z=lambda d: d["Z_Full"].abs())
    .sort_values("Abs_Z", ascending=False)
    [
        [
            "Spread",
            "Current_bp",
            "Percentile",
            "Z_Full",
            "Z_1Y",
            "AR1_phi",
            "HalfLife_obs",
            "ADF_p",
            "Persistence",
            "ADF_Stationary"
        ]
    ]
)


# ============================================================
# 7. OUTPUT
# ============================================================

print("\n================ FULL ANALYTICS ================\n")

display(
    analytics_display[
        [
            "Spread",
            "Start",
            "Obs",
            "Current_bp",
            "Mean_bp",
            "Median_bp",
            "Percentile",
            "Z_Full",
            "Z_1Y",
            "Z_63D",
            "AR1_phi",
            "HalfLife_obs",
            "ADF_p",
            "Persistence",
            "ADF_Stationary",
            "Vol_21D_bp",
            "Vol_63D_bp"
        ]
    ]
)


print("\n================ CURRENT EXTREMENESS ================\n")

display(current_extremes)


print("\n================ FIVE HIGHEST / LOWEST OBSERVATIONS ================\n")

display(
    extremes.sort_values(
        ["Spread", "Side", "Value_bp"],
        ascending=[True, True, False]
    )
)


# ============================================================
# 8. SPECIFIC SANITY CHECKS
# ============================================================

print("\n================ SANITY CHECK: SOFR-EFFR ================\n")

display(
    extremes[
        extremes["Spread"] == "SOFR-EFFR"
    ].sort_values("Value_bp", ascending=False)
)


print("\n================ SANITY CHECK: A2P2-AA NF 30D ================\n")

display(
    extremes[
        extremes["Spread"] == "A2P2-AA_NF_30D"
    ].sort_values("Value_bp")
)


print("\nDone.")
```

```python
# ============================================================
# CELL 3 — CP FUNDING-STRESS REGIME / ROLLING PERSISTENCE
# ============================================================

import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller

# ------------------------------------------------------------
# CORE CP CREDIT SPREADS
# ------------------------------------------------------------

CP = {
    "30D": spreads["A2P2-AA_NF_30D"].dropna(),
    "60D": spreads["A2P2-AA_NF_60D"].dropna(),
    "90D": spreads["A2P2-AA_NF_90D"].dropna()
}


# ------------------------------------------------------------
# ROLLING AR(1)
# ------------------------------------------------------------

def rolling_ar1(x, window=252):

    x = x.dropna().sort_index()

    phi = pd.Series(index=x.index, dtype=float)
    half_life = pd.Series(index=x.index, dtype=float)

    for i in range(window, len(x) + 1):

        sample = x.iloc[i-window:i]

        y = sample.iloc[1:].values
        lag = sample.iloc[:-1].values

        X = sm.add_constant(lag)

        try:

            model = sm.OLS(y, X).fit()

            p = model.params[1]

            phi.iloc[i-1] = p

            if 0 < p < 1:

                half_life.iloc[i-1] = (
                    -np.log(2) / np.log(p)
                )

        except:
            pass

    return phi, half_life


# ------------------------------------------------------------
# ROLLING Z-SCORE
# ------------------------------------------------------------

def rolling_z(x, window):

    mean = x.rolling(window).mean()
    std = x.rolling(window).std()

    return (x - mean) / std


# ------------------------------------------------------------
# ROLLING PERCENTILE
# ------------------------------------------------------------

def rolling_percentile(x, window=252):

    def pct(arr):

        latest = arr[-1]

        return (
            np.sum(arr <= latest)
            / len(arr)
            * 100
        )

    return x.rolling(window).apply(
        pct,
        raw=True
    )


# ------------------------------------------------------------
# ROLLING VOLATILITY OF SPREAD LEVEL
# ------------------------------------------------------------

def rolling_vol(x, window=63):

    return x.rolling(window).std()


# ------------------------------------------------------------
# BUILD ANALYTICS
# ------------------------------------------------------------

cp_analysis = {}

for maturity, x in CP.items():

    df = pd.DataFrame(index=x.index)

    df["Spread_bp"] = x

    # 1Y
    df["Z_1Y"] = rolling_z(x, 252)

    # 3Y
    df["Z_3Y"] = rolling_z(x, 756)

    # historical expanding z-score
    expanding_mean = x.expanding(
        min_periods=252
    ).mean()

    expanding_std = x.expanding(
        min_periods=252
    ).std()

    df["Z_Expanding"] = (
        (x - expanding_mean)
        / expanding_std
    )

    # Percentile
    df["Pct_1Y"] = rolling_percentile(
        x,
        252
    )

    # Volatility
    df["Vol_63D"] = rolling_vol(
        x,
        63
    )

    # Rolling persistence
    phi1, hl1 = rolling_ar1(
        x,
        252
    )

    phi3, hl3 = rolling_ar1(
        x,
        756
    )

    df["AR1_1Y"] = phi1
    df["HalfLife_1Y"] = hl1

    df["AR1_3Y"] = phi3
    df["HalfLife_3Y"] = hl3

    cp_analysis[maturity] = df


# ============================================================
# CURRENT SNAPSHOT
# ============================================================

snapshot = []

for maturity, df in cp_analysis.items():

    x = df.dropna(
        subset=["Spread_bp"]
    )

    last = x.iloc[-1]

    snapshot.append({

        "Maturity": maturity,

        "Date": x.index[-1].date(),

        "Spread_bp":
            last["Spread_bp"],

        "Z_1Y":
            last["Z_1Y"],

        "Z_3Y":
            last["Z_3Y"],

        "Z_Expanding":
            last["Z_Expanding"],

        "Pct_1Y":
            last["Pct_1Y"],

        "Vol_63D":
            last["Vol_63D"],

        "AR1_1Y":
            last["AR1_1Y"],

        "HalfLife_1Y":
            last["HalfLife_1Y"],

        "AR1_3Y":
            last["AR1_3Y"],

        "HalfLife_3Y":
            last["HalfLife_3Y"]

    })


snapshot = pd.DataFrame(snapshot)

numeric = snapshot.select_dtypes(
    include=np.number
).columns

snapshot[numeric] = (
    snapshot[numeric]
    .round(3)
)


print(
    "\n========= CURRENT CP FUNDING-STRESS SNAPSHOT =========\n"
)

display(snapshot)


# ============================================================
# REGIME CLASSIFICATION
#
# Based on 1Y z-score of credit spread.
# Purely descriptive.
# ============================================================

def regime(z):

    if pd.isna(z):
        return "N/A"

    if z >= 2:
        return "Extreme Stress"

    if z >= 1:
        return "Elevated Stress"

    if z <= -1:
        return "Compressed"

    return "Normal"


regime_table = snapshot[
    [
        "Maturity",
        "Date",
        "Spread_bp",
        "Z_1Y",
        "Pct_1Y",
        "AR1_1Y",
        "HalfLife_1Y"
    ]
].copy()

regime_table["Regime"] = (
    regime_table["Z_1Y"]
    .apply(regime)
)


print(
    "\n========= CURRENT REGIME =========\n"
)

display(regime_table)


# ============================================================
# CROSS-MATURITY COMPOSITE
#
# Standardize each maturity independently over 1Y,
# then average available z-scores.
# ============================================================

zframe = pd.concat(
    {
        maturity: df["Z_1Y"]
        for maturity, df
        in cp_analysis.items()
    },
    axis=1
)

composite = zframe.mean(
    axis=1,
    skipna=True
)

composite.name = "CP_Stress_Composite"


# Require at least 2 maturities available
valid_count = zframe.notna().sum(axis=1)

composite[
    valid_count < 2
] = np.nan


latest_composite = (
    composite
    .dropna()
    .iloc[-1]
)

print(
    "\n========= CP STRESS COMPOSITE =========\n"
)

print(
    f"Latest composite Z-score: "
    f"{latest_composite:.3f}"
)

print(
    f"Regime: "
    f"{regime(latest_composite)}"
)


# ============================================================
# RECENT HISTORY
# ============================================================

recent = pd.concat(
    [
        CP["30D"].rename("30D_bp"),
        CP["60D"].rename("60D_bp"),
        CP["90D"].rename("90D_bp"),
        composite
    ],
    axis=1
)

print(
    "\n========= LAST 20 OBSERVATIONS =========\n"
)

display(
    recent.tail(20).round(3)
)
```

```python
# ============================================================
# CELL 4 — US FUNDING STRESS EARLY-WARNING MODEL
#
# Requires Cells 1–3.
#
# 1. Creates a business-day CP composite with max 5-day carry
# 2. Adds funding/repo predictors
# 3. Tests forward CP spread widening over 5/20/60 days
# 4. Runs predictive regressions
# 5. Reports directional hit rates
#
# IMPORTANT:
# Raw series remain untouched for AR/OU estimation.
# ============================================================

import pandas as pd
import numpy as np
import statsmodels.api as sm

# ============================================================
# 1. COMMON BUSINESS-DAY CP PANEL
# ============================================================

cp_raw = pd.concat({
    "30D": CP["30D"],
    "60D": CP["60D"],
    "90D": CP["90D"]
}, axis=1).sort_index()

calendar = pd.date_range(
    cp_raw.index.min(),
    cp_raw.index.max(),
    freq="B"
)

cp_aligned = cp_raw.reindex(calendar)

# Carry latest observation for MAXIMUM 5 business days.
# This is ONLY for composite construction.
cp_filled = cp_aligned.ffill(limit=5)


# ============================================================
# 2. ROLLING Z-SCORES
#
# Each maturity standardized independently using trailing
# 252 business-day observations.
# ============================================================

cp_z = pd.DataFrame(index=cp_filled.index)

for col in cp_filled.columns:

    x = cp_filled[col]

    mean = x.rolling(
        252,
        min_periods=126
    ).mean()

    std = x.rolling(
        252,
        min_periods=126
    ).std()

    cp_z[col] = (x - mean) / std


# Require at least 2 maturities
valid_count = cp_z.notna().sum(axis=1)

cp_composite = cp_z.mean(
    axis=1,
    skipna=True
)

cp_composite[
    valid_count < 2
] = np.nan

cp_composite.name = "CP_Stress"


# ============================================================
# 3. CP TERM-STRUCTURE FACTOR
#
# Average 90D minus 30D spread.
# Positive = longer CP funding more expensive.
# ============================================================

cp_term = (
    cp_filled["90D"]
    - cp_filled["30D"]
)

cp_term.name = "CP_Term_90D_30D"


# ============================================================
# 4. FUNDING-MARKET PREDICTORS
#
# Use existing spreads from Cell 1.
# Align onto business-day calendar.
# Small 5-day carry only for predictor panel.
# ============================================================

predictor_raw = pd.concat({

    "SOFR_EFFR":
        spreads["SOFR-EFFR"],

    "TGCR_SOFR":
        spreads["TGCR-SOFR"],

    "BGCR_TGCR":
        spreads["BGCR-TGCR"],

    "SOFR_IORB":
        spreads["SOFR-IORB"],

    "EFFR_IORB":
        spreads["EFFR-IORB"],

    "SOFR_RRP":
        spreads["SOFR-ON_RRP"]

}, axis=1).sort_index()


predictors = (
    predictor_raw
    .reindex(calendar)
    .ffill(limit=5)
)


# ============================================================
# 5. STANDARDIZE PREDICTORS
#
# Trailing 252 business days.
# This puts everything on comparable scale.
# ============================================================

predictor_z = pd.DataFrame(
    index=predictors.index
)

for col in predictors.columns:

    x = predictors[col]

    mean = x.rolling(
        252,
        min_periods=126
    ).mean()

    std = x.rolling(
        252,
        min_periods=126
    ).std()

    z = (x - mean) / std

    # Avoid division-by-zero artefacts
    z = z.replace(
        [np.inf, -np.inf],
        np.nan
    )

    predictor_z[col] = z


# ============================================================
# 6. MASTER DATASET
# ============================================================

model = pd.DataFrame(
    index=calendar
)

model["CP_Stress"] = cp_composite
model["CP_Term"] = cp_term

for col in predictor_z.columns:

    model[col] = predictor_z[col]


# ============================================================
# 7. ADD CP STATE VARIABLES
# ============================================================

# 20-day change in stress
model["CP_Momentum_20D"] = (
    model["CP_Stress"]
    - model["CP_Stress"].shift(20)
)

# 20-day volatility of daily CP composite changes
model["CP_Vol_20D"] = (
    model["CP_Stress"]
    .diff()
    .rolling(20)
    .std()
)

# 60-day volatility
model["CP_Vol_60D"] = (
    model["CP_Stress"]
    .diff()
    .rolling(60)
    .std()
)


# ============================================================
# 8. FORWARD TARGETS
#
# Positive = CP stress widens in future.
# ============================================================

HORIZONS = [5, 20, 60]

for h in HORIZONS:

    model[f"Fwd_CP_{h}D"] = (
        model["CP_Stress"].shift(-h)
        - model["CP_Stress"]
    )


# ============================================================
# 9. PREDICTOR SET
# ============================================================

FEATURES = [

    "CP_Stress",
    "CP_Term",

    "CP_Momentum_20D",
    "CP_Vol_20D",
    "CP_Vol_60D",

    "SOFR_EFFR",
    "TGCR_SOFR",
    "BGCR_TGCR",

    "SOFR_IORB",
    "EFFR_IORB",
    "SOFR_RRP"
]


# ============================================================
# 10. UNIVARIATE PREDICTIVE REGRESSIONS
#
# Future CP widening = a + b * predictor_t
#
# HAC standard errors because forward returns overlap.
# ============================================================

uni_results = []

for h in HORIZONS:

    target = f"Fwd_CP_{h}D"

    for feature in FEATURES:

        sample = model[
            [feature, target]
        ].dropna()

        if len(sample) < 100:
            continue

        X = sm.add_constant(
            sample[feature]
        )

        y = sample[target]

        fit = sm.OLS(
            y,
            X
        ).fit(
            cov_type="HAC",
            cov_kwds={
                "maxlags": h
            }
        )

        uni_results.append({

            "Horizon": f"{h}D",

            "Predictor": feature,

            "N": len(sample),

            "Beta": fit.params[feature],

            "t_stat": fit.tvalues[feature],

            "p_value": fit.pvalues[feature],

            "R2": fit.rsquared

        })


uni_results = pd.DataFrame(
    uni_results
)

uni_results[
    ["Beta", "t_stat", "p_value", "R2"]
] = uni_results[
    ["Beta", "t_stat", "p_value", "R2"]
].round(4)


# ============================================================
# 11. MULTIVARIATE MODEL
#
# IMPORTANT:
# Different predictors start at different dates.
#
# Therefore:
# We DON'T force IORB predictors into earlier samples.
#
# Build:
#   A) Core model — available from SOFR era
#   B) IORB model — post-2021 only
# ============================================================

CORE_FEATURES = [

    "CP_Stress",
    "CP_Term",

    "CP_Momentum_20D",
    "CP_Vol_20D",

    "SOFR_EFFR",
    "TGCR_SOFR",
    "SOFR_RRP"
]


IORB_FEATURES = [

    "CP_Stress",
    "CP_Term",

    "CP_Momentum_20D",
    "CP_Vol_20D",

    "SOFR_EFFR",
    "TGCR_SOFR",

    "SOFR_IORB",
    "EFFR_IORB",
    "SOFR_RRP"
]


multi_results = []


def run_multivariate(features, label, h):

    target = f"Fwd_CP_{h}D"

    sample = model[
        features + [target]
    ].dropna()

    if len(sample) < 100:
        return

    X = sm.add_constant(
        sample[features]
    )

    y = sample[target]

    fit = sm.OLS(
        y,
        X
    ).fit(
        cov_type="HAC",
        cov_kwds={
            "maxlags": h
        }
    )

    for variable in features:

        multi_results.append({

            "Model": label,

            "Horizon": f"{h}D",

            "Variable": variable,

            "N": len(sample),

            "Beta":
                fit.params[variable],

            "t_stat":
                fit.tvalues[variable],

            "p_value":
                fit.pvalues[variable],

            "Model_R2":
                fit.rsquared,

            "Adj_R2":
                fit.rsquared_adj

        })


for h in HORIZONS:

    run_multivariate(
        CORE_FEATURES,
        "CORE",
        h
    )

    run_multivariate(
        IORB_FEATURES,
        "IORB",
        h
    )


multi_results = pd.DataFrame(
    multi_results
)

multi_results[
    [
        "Beta",
        "t_stat",
        "p_value",
        "Model_R2",
        "Adj_R2"
    ]
] = multi_results[
    [
        "Beta",
        "t_stat",
        "p_value",
        "Model_R2",
        "Adj_R2"
    ]
].round(4)


# ============================================================
# 12. SIMPLE DIRECTIONAL TEST
#
# Does predictor being HIGH (> +1z) precede widening more
# frequently than predictor being LOW (< -1z)?
#
# No fitted model involved.
# ============================================================

directional = []

for h in HORIZONS:

    target = f"Fwd_CP_{h}D"

    for feature in FEATURES:

        sample = model[
            [feature, target]
        ].dropna()

        if len(sample) < 100:
            continue

        high = sample[
            sample[feature] >= 1
        ]

        low = sample[
            sample[feature] <= -1
        ]

        high_widen = (
            (high[target] > 0).mean()
            if len(high)
            else np.nan
        )

        low_widen = (
            (low[target] > 0).mean()
            if len(low)
            else np.nan
        )

        directional.append({

            "Horizon": f"{h}D",

            "Predictor": feature,

            "High_N": len(high),

            "P_Widen_if_High":
                high_widen,

            "Low_N": len(low),

            "P_Widen_if_Low":
                low_widen,

            "Difference":
                high_widen - low_widen

        })


directional = pd.DataFrame(
    directional
)

for col in [
    "P_Widen_if_High",
    "P_Widen_if_Low",
    "Difference"
]:

    directional[col] = (
        directional[col] * 100
    ).round(2)


# ============================================================
# 13. CURRENT CONDITIONS
# ============================================================

current_cols = [

    "CP_Stress",
    "CP_Term",
    "CP_Momentum_20D",
    "CP_Vol_20D",

    "SOFR_EFFR",
    "TGCR_SOFR",
    "BGCR_TGCR",

    "SOFR_IORB",
    "EFFR_IORB",
    "SOFR_RRP"
]

latest = (
    model[current_cols]
    .ffill()
    .iloc[-1]
    .to_frame("Current")
)

latest["Current"] = (
    latest["Current"]
    .round(3)
)


# ============================================================
# 14. OUTPUT
# ============================================================

print(
    "\n========== CURRENT FUNDING CONDITIONS ==========\n"
)

display(latest)


print(
    "\n========== UNIVARIATE PREDICTIVE TESTS ==========\n"
)

display(
    uni_results.sort_values(
        ["Horizon", "p_value"]
    )
)


print(
    "\n========== MULTIVARIATE MODELS ==========\n"
)

display(
    multi_results.sort_values(
        ["Model", "Horizon", "p_value"]
    )
)


print(
    "\n========== HIGH-vs-LOW DIRECTIONAL TEST ==========\n"
)

display(
    directional.sort_values(
        ["Horizon", "Difference"],
        ascending=[True, False]
    )
)


print("\nDone.")
```

```python
# ============================================================
# BUILD + PUSH STREAMLIT APP
# U.S. Funding Conditions
# ============================================================

import os
import subprocess
from getpass import getpass

REPO_DIR = "/content/U.S.-Funding-Conditions"

# ============================================================
# APP.PY
# ============================================================

app_code = r'''
import streamlit as st
import pandas as pd
import numpy as np
import json
import plotly.graph_objects as go
from pathlib import Path


# ============================================================
# CONFIG
# ============================================================

st.set_page_config(
    page_title="U.S. Funding Conditions",
    page_icon="🏦",
    layout="wide",
    initial_sidebar_state="expanded"
)

DATA = Path("data")


# ============================================================
# STYLE
# ============================================================

st.markdown("""
<style>

.block-container {
    padding-top: 1.5rem;
    padding-bottom: 3rem;
}

[data-testid="stMetricValue"] {
    font-size: 1.65rem;
}

[data-testid="stMetricLabel"] {
    font-size: 0.90rem;
}

.small-note {
    font-size: 0.82rem;
    color: #888;
}

.section-note {
    font-size: 0.88rem;
    color: #888;
    margin-bottom: 0.8rem;
}

</style>
""", unsafe_allow_html=True)


# ============================================================
# DATA LOADERS
# ============================================================

@st.cache_data
def load_data():

    rates = pd.read_csv(
        DATA / "raw_rates.csv",
        index_col=0,
        parse_dates=True
    )

    spreads = pd.read_csv(
        DATA / "spreads.csv",
        index_col=0,
        parse_dates=True
    )

    cp = pd.read_csv(
        DATA / "cp_monitor.csv",
        index_col=0,
        parse_dates=True
    )

    latest_rates = pd.read_csv(
        DATA / "latest_rates.csv"
    )

    latest_spreads = pd.read_csv(
        DATA / "latest_spreads.csv"
    )

    latest_cp = pd.read_csv(
        DATA / "latest_cp.csv"
    )

    quality = pd.read_csv(
        DATA / "data_quality.csv"
    )

    with open(DATA / "metadata.json", "r") as f:
        metadata = json.load(f)

    return (
        rates,
        spreads,
        cp,
        latest_rates,
        latest_spreads,
        latest_cp,
        quality,
        metadata
    )


(
    rates,
    spreads,
    cp,
    latest_rates,
    latest_spreads,
    latest_cp,
    quality,
    metadata
) = load_data()


# ============================================================
# HELPERS
# ============================================================

def line_chart(
    df,
    columns,
    title,
    ytitle,
    zero_line=False
):

    fig = go.Figure()

    for col in columns:

        if col not in df.columns:
            continue

        x = df[col].dropna()

        fig.add_trace(
            go.Scatter(
                x=x.index,
                y=x.values,
                mode="lines",
                name=col,
                line=dict(width=1.6)
            )
        )

    if zero_line:
        fig.add_hline(
            y=0,
            line_dash="dot",
            opacity=0.5
        )

    fig.update_layout(
        title=title,
        yaxis_title=ytitle,
        xaxis_title="",
        hovermode="x unified",
        height=480,
        margin=dict(l=20, r=20, t=55, b=20),
        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=1.02,
            xanchor="left",
            x=0
        )
    )

    return fig


def get_rate(name):

    row = latest_rates[
        latest_rates["Series"] == name
    ]

    if row.empty:
        return np.nan, None

    return (
        float(row.iloc[0]["Current"]),
        str(row.iloc[0]["Date"])
    )


def get_spread(name):

    row = latest_spreads[
        latest_spreads["Spread"] == name
    ]

    if row.empty:
        return None

    return row.iloc[0]


def regime_from_composite(z):

    if pd.isna(z):
        return "N/A"

    if z >= 2:
        return "Extreme Stress"

    if z >= 1:
        return "Elevated Stress"

    if z <= -1:
        return "Compressed"

    return "Normal"


# ============================================================
# SIDEBAR
# ============================================================

st.sidebar.title("U.S. Funding Conditions")

page = st.sidebar.radio(
    "Section",
    [
        "Overview",
        "Funding Basis",
        "Commercial Paper",
        "Persistence",
        "Data & Methodology"
    ]
)

st.sidebar.divider()

st.sidebar.caption(
    "Federal Reserve Bank of New York / "
    "Federal Reserve / FRED"
)


# ============================================================
# HEADER
# ============================================================

st.title("U.S. Funding Conditions")

st.caption(
    "Short-term funding, repo-market basis, commercial-paper "
    "credit conditions and persistence."
)


# ============================================================
# OVERVIEW
# ============================================================

if page == "Overview":

    # --------------------------------------------------------
    # CURRENT POLICY / FUNDING RATES
    # --------------------------------------------------------

    st.subheader("Current Funding Rates")

    names = [
        "SOFR",
        "EFFR",
        "IORB",
        "ON_RRP",
        "TGCR",
        "BGCR"
    ]

    cols = st.columns(6)

    for c, name in zip(cols, names):

        value, date = get_rate(name)

        with c:

            st.metric(
                name.replace("_", " "),
                f"{value:.2f}%"
                if pd.notna(value)
                else "N/A"
            )

            if date:
                st.caption(date)


    # --------------------------------------------------------
    # KEY BASIS
    # --------------------------------------------------------

    st.divider()
    st.subheader("Key Funding Basis")

    key_spreads = [
        "SOFR-EFFR",
        "TGCR-SOFR",
        "SOFR-IORB",
        "EFFR-IORB",
        "SOFR-ON_RRP"
    ]

    cols = st.columns(len(key_spreads))

    for c, name in zip(cols, key_spreads):

        row = get_spread(name)

        with c:

            if row is not None:

                st.metric(
                    name,
                    f"{row['Current_bp']:.1f} bp"
                )

                st.caption(
                    f"1Y z: {row['Z_1Y']:.2f}"
                )


    # --------------------------------------------------------
    # CP STRESS
    # --------------------------------------------------------

    st.divider()
    st.subheader("Commercial-Paper Credit Conditions")

    composite = (
        cp["CP_Stress_Composite"]
        .dropna()
    )

    latest_composite = composite.iloc[-1]
    latest_date = composite.index[-1]

    cp30 = latest_cp[
        latest_cp["Maturity"] == "30D"
    ].iloc[0]

    cp60 = latest_cp[
        latest_cp["Maturity"] == "60D"
    ].iloc[0]

    cp90 = latest_cp[
        latest_cp["Maturity"] == "90D"
    ].iloc[0]

    c1, c2, c3, c4 = st.columns(4)

    c1.metric(
        "CP Stress Composite",
        f"{latest_composite:.2f}σ"
    )

    c1.caption(
        regime_from_composite(
            latest_composite
        )
    )

    c2.metric(
        "30D A2/P2 − AA",
        f"{cp30['Spread_bp']:.0f} bp"
    )

    c2.caption(
        f"1Y z: {cp30['Z_1Y']:.2f}"
    )

    c3.metric(
        "60D A2/P2 − AA",
        f"{cp60['Spread_bp']:.0f} bp"
    )

    c3.caption(
        f"1Y z: {cp60['Z_1Y']:.2f}"
    )

    c4.metric(
        "90D A2/P2 − AA",
        f"{cp90['Spread_bp']:.0f} bp"
    )

    c4.caption(
        f"1Y z: {cp90['Z_1Y']:.2f}"
    )


    # --------------------------------------------------------
    # COMPOSITE CHART
    # --------------------------------------------------------

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=composite.index,
            y=composite.values,
            mode="lines",
            name="CP Stress Composite"
        )
    )

    fig.add_hline(
        y=0,
        line_dash="dot",
        opacity=0.5
    )

    fig.add_hline(
        y=1,
        line_dash="dash",
        opacity=0.35
    )

    fig.add_hline(
        y=-1,
        line_dash="dash",
        opacity=0.35
    )

    fig.update_layout(
        title="Commercial-Paper Funding Stress Composite",
        yaxis_title="Z-score",
        xaxis_title="",
        hovermode="x unified",
        height=500,
        margin=dict(l=20, r=20, t=55, b=20)
    )

    st.plotly_chart(
        fig,
        use_container_width=True
    )

    st.caption(
        f"Latest composite observation: "
        f"{latest_date.date()}"
    )


# ============================================================
# FUNDING BASIS
# ============================================================

elif page == "Funding Basis":

    st.subheader("Funding & Repo Basis")

    basis_groups = {

        "Overnight / Repo": [
            "SOFR-EFFR",
            "TGCR-SOFR",
            "BGCR-TGCR"
        ],

        "Administered Rate Basis": [
            "SOFR-IORB",
            "EFFR-IORB",
            "SOFR-ON_RRP"
        ]
    }

    group = st.radio(
        "Basis family",
        list(basis_groups.keys()),
        horizontal=True
    )

    selected = st.multiselect(
        "Series",
        basis_groups[group],
        default=basis_groups[group]
    )

    st.plotly_chart(
        line_chart(
            spreads,
            selected,
            group,
            "Basis points",
            zero_line=True
        ),
        use_container_width=True
    )


    # --------------------------------------------------------
    # CURRENT ANALYTICS TABLE
    # --------------------------------------------------------

    table = latest_spreads[
        latest_spreads["Spread"].isin(
            basis_groups[group]
        )
    ].copy()

    cols = [
        "Spread",
        "End",
        "Current_bp",
        "Percentile",
        "Z_Full",
        "Z_1Y",
        "AR1_phi",
        "HalfLife_obs",
        "ADF_p",
        "Regime"
    ]

    table = table[cols]

    table.columns = [
        "Spread",
        "Date",
        "Current (bp)",
        "Hist. Percentile",
        "Full Z",
        "1Y Z",
        "AR(1)",
        "Half-Life",
        "ADF p",
        "Regime"
    ]

    st.dataframe(
        table,
        use_container_width=True,
        hide_index=True
    )


# ============================================================
# COMMERCIAL PAPER
# ============================================================

elif page == "Commercial Paper":

    st.subheader("Commercial-Paper Credit")

    cp_credit_cols = [
        "A2P2-AA_NF_30D",
        "A2P2-AA_NF_60D",
        "A2P2-AA_NF_90D"
    ]

    st.plotly_chart(
        line_chart(
            spreads,
            cp_credit_cols,
            "A2/P2 − AA Nonfinancial Commercial Paper",
            "Basis points",
            zero_line=True
        ),
        use_container_width=True
    )


    # --------------------------------------------------------
    # CURRENT CP TABLE
    # --------------------------------------------------------

    display_cp = latest_cp.copy()

    display_cp = display_cp[
        [
            "Maturity",
            "Date",
            "Spread_bp",
            "Z_1Y",
            "Z_3Y",
            "Pct_1Y",
            "Vol_63D",
            "AR1_1Y",
            "HalfLife_1Y",
            "Regime"
        ]
    ]

    display_cp.columns = [
        "Maturity",
        "Date",
        "Spread (bp)",
        "1Y Z",
        "3Y Z",
        "1Y Percentile",
        "63D Vol",
        "1Y AR(1)",
        "1Y Half-Life",
        "Regime"
    ]

    st.dataframe(
        display_cp,
        use_container_width=True,
        hide_index=True
    )


    # --------------------------------------------------------
    # OTHER CP RELATIONSHIPS
    # --------------------------------------------------------

    st.divider()
    st.subheader("Sector & Term Structure")

    option = st.radio(
        "View",
        [
            "AA Financial − AA Nonfinancial",
            "CP Term Structure"
        ],
        horizontal=True
    )

    if option == "AA Financial − AA Nonfinancial":

        cols = [
            "AA_FIN-AA_NF_30D",
            "AA_FIN-AA_NF_60D",
            "AA_FIN-AA_NF_90D"
        ]

    else:

        cols = [
            "AA_FIN_90D-30D",
            "AA_NF_90D-30D",
            "A2P2_NF_90D-30D"
        ]

    st.plotly_chart(
        line_chart(
            spreads,
            cols,
            option,
            "Basis points",
            zero_line=True
        ),
        use_container_width=True
    )


# ============================================================
# PERSISTENCE
# ============================================================

elif page == "Persistence":

    st.subheader("Commercial-Paper Persistence")

    maturity = st.selectbox(
        "Maturity",
        ["30D", "60D", "90D"]
    )

    ar_col = f"CP_{maturity}_AR1_1Y"
    hl_col = f"CP_{maturity}_HalfLife_1Y"

    ar3_col = f"CP_{maturity}_AR1_3Y"
    hl3_col = f"CP_{maturity}_HalfLife_3Y"


    # --------------------------------------------------------
    # CURRENT METRICS
    # --------------------------------------------------------

    row = latest_cp[
        latest_cp["Maturity"] == maturity
    ].iloc[0]

    c1, c2, c3, c4 = st.columns(4)

    c1.metric(
        "1Y AR(1)",
        f"{row['AR1_1Y']:.3f}"
    )

    c2.metric(
        "1Y Half-Life",
        f"{row['HalfLife_1Y']:.2f} obs."
    )

    c3.metric(
        "3Y AR(1)",
        f"{row['AR1_3Y']:.3f}"
    )

    c4.metric(
        "3Y Half-Life",
        f"{row['HalfLife_3Y']:.2f} obs."
    )


    # --------------------------------------------------------
    # AR CHART
    # --------------------------------------------------------

    st.plotly_chart(
        line_chart(
            cp,
            [ar_col, ar3_col],
            f"{maturity} Rolling AR(1)",
            "AR(1)",
            zero_line=False
        ),
        use_container_width=True
    )


    # --------------------------------------------------------
    # HALF LIFE CHART
    # --------------------------------------------------------

    st.plotly_chart(
        line_chart(
            cp,
            [hl_col, hl3_col],
            f"{maturity} Rolling Mean-Reversion Half-Life",
            "Observations",
            zero_line=False
        ),
        use_container_width=True
    )

    st.caption(
        "Half-life is reported only where 0 < AR(1) < 1."
    )


# ============================================================
# DATA & METHODOLOGY
# ============================================================

elif page == "Data & Methodology":

    st.subheader("Data Coverage")

    quality_display = quality[
        [
            "Series",
            "Start",
            "End",
            "Observations"
        ]
    ].copy()

    st.dataframe(
        quality_display,
        use_container_width=True,
        hide_index=True
    )

    st.divider()

    st.subheader("Methodology")

    methodology = metadata.get(
        "methodology",
        {}
    )

    for key, value in methodology.items():

        st.markdown(
            f"**{key.replace('_', ' ').title()}**"
        )

        st.write(value)

    st.divider()

    st.subheader("Sources")

    sources = metadata.get(
        "sources",
        {}
    )

    for key, value in sources.items():

        st.write(
            f"**{key.replace('_', ' ')}:** {value}"
        )

    st.divider()

    st.caption(
        "This application is an analytical market-monitoring "
        "tool and does not constitute investment advice."
    )
'''

with open(
    os.path.join(REPO_DIR, "app.py"),
    "w"
) as f:
    f.write(app_code)


# ============================================================
# REQUIREMENTS
# ============================================================

requirements = """streamlit
pandas
numpy
plotly
pyarrow
"""

with open(
    os.path.join(REPO_DIR, "requirements.txt"),
    "w"
) as f:
    f.write(requirements)


# ============================================================
# QUICK SYNTAX CHECK
# ============================================================

subprocess.run(
    ["python", "-m", "py_compile", "app.py"],
    cwd=REPO_DIR,
    check=True
)

print("app.py syntax: OK")


# ============================================================
# GIT STATUS
# ============================================================

print("\nGit status:\n")

subprocess.run(
    ["git", "status", "--short"],
    cwd=REPO_DIR,
    check=True
)


# ============================================================
# COMMIT
# ============================================================

subprocess.run(
    ["git", "add", "app.py", "requirements.txt"],
    cwd=REPO_DIR,
    check=True
)

subprocess.run(
    [
        "git",
        "commit",
        "-m",
        "Add Streamlit dashboard"
    ],
    cwd=REPO_DIR,
    check=True
)


# ============================================================
# AUTHENTICATE + PUSH
# ============================================================

token = getpass(
    "\nPaste GitHub PAT: "
)

auth_url = (
    f"https://YieldShock13:{token}"
    "@github.com/YieldShock13/"
    "U.S.-Funding-Conditions.git"
)

subprocess.run(
    ["git", "remote", "set-url", "origin", auth_url],
    cwd=REPO_DIR,
    check=True
)

subprocess.run(
    ["git", "push", "origin", "main"],
    cwd=REPO_DIR,
    check=True
)


# ============================================================
# REMOVE TOKEN FROM REMOTE
# ============================================================

subprocess.run(
    [
        "git",
        "remote",
        "set-url",
        "origin",
        "https://github.com/YieldShock13/"
        "U.S.-Funding-Conditions.git"
    ],
    cwd=REPO_DIR,
    check=True
)

token = None

print("\n========================================")
print("STREAMLIT APP PUSHED TO GITHUB")
print("========================================")
print("app.py")
print("requirements.txt")
```

```python
# ============================================================
# V2 CORRECTION + AUDIT
# U.S. FUNDING CONDITIONS
#
# FIXES:
# - 1Y = trailing 1 calendar year, NOT 252 observations
# - 3Y = trailing 3 calendar years, NOT 756 observations
# - Raw statistical estimation remains unfilled
# - CP monitoring composite may carry max 5 business days
# - AR(1) half-life explicitly = observations
# - Adds 1Y percentile + full-history percentile
# - Rebuilds CP composite
# - Validates spread arithmetic independently
#
# DOES NOT OVERWRITE V1
# ============================================================

import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller
import os
import json

OUT2 = "funding_monitor_v2_audit"
os.makedirs(OUT2, exist_ok=True)

# ------------------------------------------------------------
# REQUIRE EXISTING OBJECTS
# ------------------------------------------------------------

for obj in ["rates", "spreads"]:
    if obj not in globals():
        raise RuntimeError(
            f"{obj} not found. Run the original data-download/"
            "construction cells first."
        )

rates2 = rates.copy().sort_index()

# ============================================================
# 1. HELPERS — TRUE CALENDAR WINDOWS
# ============================================================

def calendar_window(x, years):
    x = x.dropna().sort_index()

    if x.empty:
        return x

    end = x.index[-1]
    start = end - pd.DateOffset(years=years)

    return x.loc[x.index >= start]


def zscore_current(x):
    x = x.dropna()

    if len(x) < 2:
        return np.nan

    sd = x.std()

    if pd.isna(sd) or sd == 0:
        return np.nan

    return (x.iloc[-1] - x.mean()) / sd


def percentile_current(x):
    x = x.dropna()

    if x.empty:
        return np.nan

    return (x <= x.iloc[-1]).mean() * 100


def ar1_stats(x):
    """
    AR(1):
        x_t = c + phi*x_(t-1) + e_t

    Half-life:
        -ln(2)/ln(phi)

    IMPORTANT:
    Half-life is in OBSERVATIONS, not calendar days.
    """

    x = x.dropna().astype(float)

    if len(x) < 20:
        return np.nan, np.nan

    y = x.iloc[1:].values
    lag = x.iloc[:-1].values

    X = sm.add_constant(lag)

    try:
        fit = sm.OLS(y, X).fit()
        phi = float(fit.params[1])
    except Exception:
        return np.nan, np.nan

    if 0 < phi < 1:
        hl = -np.log(2) / np.log(phi)
    else:
        hl = np.nan

    return phi, hl


def adf_pvalue(x):
    x = x.dropna().astype(float)

    if len(x) < 30:
        return np.nan

    try:
        return adfuller(
            x,
            regression="c",
            autolag="AIC"
        )[1]
    except Exception:
        return np.nan


def annualized_obs_count(x):
    """
    Diagnostic only:
    actual observations contained in trailing calendar year.
    """
    return len(calendar_window(x, 1))


# ============================================================
# 2. REBUILD / VERIFY SPREAD DEFINITIONS
# ============================================================

spread_defs = {
    "SOFR-EFFR": ("SOFR", "EFFR"),
    "TGCR-SOFR": ("TGCR", "SOFR"),
    "BGCR-TGCR": ("BGCR", "TGCR"),
    "SOFR-IORB": ("SOFR", "IORB"),
    "EFFR-IORB": ("EFFR", "IORB"),
    "SOFR-ON_RRP": ("SOFR", "ON_RRP"),

    "A2P2-AA_NF_30D": ("A2P2_NF_30D", "AA_NF_30D"),
    "A2P2-AA_NF_60D": ("A2P2_NF_60D", "AA_NF_60D"),
    "A2P2-AA_NF_90D": ("A2P2_NF_90D", "AA_NF_90D"),

    "AA_FIN-AA_NF_30D": ("AA_FIN_30D", "AA_NF_30D"),
    "AA_FIN-AA_NF_60D": ("AA_FIN_60D", "AA_NF_60D"),
    "AA_FIN-AA_NF_90D": ("AA_FIN_90D", "AA_NF_90D"),

    "AA_FIN_90D-30D": ("AA_FIN_90D", "AA_FIN_30D"),
    "AA_NF_90D-30D": ("AA_NF_90D", "AA_NF_30D"),
    "A2P2_NF_90D-30D": ("A2P2_NF_90D", "A2P2_NF_30D"),
}

spreads2 = {}

for name, (lhs, rhs) in spread_defs.items():

    pair = pd.concat(
        [rates2[lhs], rates2[rhs]],
        axis=1,
        join="inner"
    ).dropna()

    # Raw rates are percentage points.
    # Difference ×100 converts to basis points.
    spreads2[name] = (
        pair.iloc[:, 0] - pair.iloc[:, 1]
    ) * 100

spreads2 = pd.DataFrame(spreads2).sort_index()
spreads2.index.name = "Date"


# ============================================================
# 3. INDEPENDENT SPREAD-ARITHMETIC AUDIT
# ============================================================

spread_audit = []

for name in spread_defs:

    old = spreads[name].dropna().sort_index()
    new = spreads2[name].dropna().sort_index()

    common = old.index.intersection(new.index)

    if len(common):

        diff = (
            old.loc[common] -
            new.loc[common]
        ).abs()

        max_error = diff.max()

    else:
        max_error = np.nan

    spread_audit.append({
        "Spread": name,
        "Old_N": len(old),
        "Rebuilt_N": len(new),
        "Common_N": len(common),
        "Max_Abs_Error_bp": max_error,
        "PASS": (
            pd.notna(max_error)
            and max_error < 1e-8
        )
    })

spread_audit = pd.DataFrame(spread_audit)


# ============================================================
# 4. CORRECTED SPREAD ANALYTICS
# ============================================================

analytics2 = []

for name in spreads2.columns:

    x = spreads2[name].dropna()

    if x.empty:
        continue

    x1 = calendar_window(x, 1)
    x3 = calendar_window(x, 3)

    phi_full, hl_full = ar1_stats(x)
    phi_1y, hl_1y = ar1_stats(x1)
    phi_3y, hl_3y = ar1_stats(x3)

    analytics2.append({

        "Spread": name,

        "Start":
            x.index[0].date(),

        "End":
            x.index[-1].date(),

        "Current_bp":
            x.iloc[-1],

        "Observations":
            len(x),

        # Full history
        "Full_Percentile":
            percentile_current(x),

        "Z_Full":
            zscore_current(x),

        "AR1_Full":
            phi_full,

        "HalfLife_Full_obs":
            hl_full,

        "ADF_p":
            adf_pvalue(x),

        # TRUE trailing calendar year
        "Start_1Y":
            x1.index[0].date()
            if len(x1) else None,

        "Obs_1Y":
            len(x1),

        "Pct_1Y":
            percentile_current(x1),

        "Z_1Y":
            zscore_current(x1),

        "AR1_1Y":
            phi_1y,

        "HalfLife_1Y_obs":
            hl_1y,

        # TRUE trailing 3 calendar years
        "Start_3Y":
            x3.index[0].date()
            if len(x3) else None,

        "Obs_3Y":
            len(x3),

        "Pct_3Y":
            percentile_current(x3),

        "Z_3Y":
            zscore_current(x3),

        "AR1_3Y":
            phi_3y,

        "HalfLife_3Y_obs":
            hl_3y
    })


analytics2 = pd.DataFrame(analytics2)


# ============================================================
# 5. REGIME — BASED ON TRUE 1Y Z
# ============================================================

def generic_regime(z):

    if pd.isna(z):
        return np.nan

    if z >= 2:
        return "Very High"

    if z >= 1:
        return "High"

    if z <= -2:
        return "Very Low"

    if z <= -1:
        return "Low"

    return "Normal"


analytics2["Regime"] = (
    analytics2["Z_1Y"]
    .apply(generic_regime)
)


# ============================================================
# 6. RAW CP CREDIT SPREADS
#
# NO FORWARD FILL HERE.
# ============================================================

CP2 = {
    "30D": spreads2["A2P2-AA_NF_30D"].dropna(),
    "60D": spreads2["A2P2-AA_NF_60D"].dropna(),
    "90D": spreads2["A2P2-AA_NF_90D"].dropna()
}


# ============================================================
# 7. CORRECTED CURRENT CP ANALYTICS
# ============================================================

cp_rows = []

for maturity, x in CP2.items():

    x = x.dropna().sort_index()

    x1 = calendar_window(x, 1)
    x3 = calendar_window(x, 3)

    phi1, hl1 = ar1_stats(x1)
    phi3, hl3 = ar1_stats(x3)

    expanding_z = zscore_current(x)

    # Keep 63-observation volatility explicitly labelled OBS.
    # We are NOT calling this "3 months".
    x63 = x.iloc[-63:]

    vol63 = (
        x63.std()
        if len(x63) >= 2
        else np.nan
    )

    cp_rows.append({

        "Maturity":
            maturity,

        "Date":
            x.index[-1].date(),

        "Spread_bp":
            x.iloc[-1],

        "Start_1Y":
            x1.index[0].date(),

        "Obs_1Y":
            len(x1),

        "Z_1Y":
            zscore_current(x1),

        "Pct_1Y":
            percentile_current(x1),

        "AR1_1Y":
            phi1,

        "HalfLife_1Y_obs":
            hl1,

        "Start_3Y":
            x3.index[0].date(),

        "Obs_3Y":
            len(x3),

        "Z_3Y":
            zscore_current(x3),

        "Pct_3Y":
            percentile_current(x3),

        "AR1_3Y":
            phi3,

        "HalfLife_3Y_obs":
            hl3,

        "Z_Expanding":
            expanding_z,

        "Vol_63obs":
            vol63
    })


cp_snapshot2 = pd.DataFrame(cp_rows)


def cp_regime(z):

    if pd.isna(z):
        return np.nan

    if z >= 2:
        return "Extreme Stress"

    if z >= 1:
        return "Elevated Stress"

    if z <= -1:
        return "Compressed"

    return "Normal"


cp_snapshot2["Regime"] = (
    cp_snapshot2["Z_1Y"]
    .apply(cp_regime)
)


# ============================================================
# 8. HISTORICAL ROLLING CALENDAR-WINDOW ANALYTICS
#
# IMPORTANT:
# For each observation date t:
#   1Y = t minus 1 calendar year
#   3Y = t minus 3 calendar years
#
# No forward-fill during estimation.
# ============================================================

def rolling_calendar_metrics(x):

    x = x.dropna().sort_index()

    out = pd.DataFrame(index=x.index)

    z1 = []
    z3 = []
    pct1 = []
    pct3 = []

    ar1_1 = []
    hl1 = []

    ar1_3 = []
    hl3 = []

    obs1 = []
    obs3 = []

    for dt in x.index:

        hist = x.loc[:dt]

        w1 = hist.loc[
            hist.index >= dt - pd.DateOffset(years=1)
        ]

        w3 = hist.loc[
            hist.index >= dt - pd.DateOffset(years=3)
        ]

        z1.append(zscore_current(w1))
        z3.append(zscore_current(w3))

        pct1.append(percentile_current(w1))
        pct3.append(percentile_current(w3))

        p1, h1 = ar1_stats(w1)
        p3, h3 = ar1_stats(w3)

        ar1_1.append(p1)
        hl1.append(h1)

        ar1_3.append(p3)
        hl3.append(h3)

        obs1.append(len(w1))
        obs3.append(len(w3))

    out["Z_1Y"] = z1
    out["Z_3Y"] = z3

    out["Pct_1Y"] = pct1
    out["Pct_3Y"] = pct3

    out["AR1_1Y"] = ar1_1
    out["HalfLife_1Y_obs"] = hl1

    out["AR1_3Y"] = ar1_3
    out["HalfLife_3Y_obs"] = hl3

    out["Obs_1Y"] = obs1
    out["Obs_3Y"] = obs3

    return out


cp_analysis2 = {}

for maturity, x in CP2.items():

    print(
        f"Calculating corrected rolling metrics: "
        f"{maturity}..."
    )

    cp_analysis2[maturity] = (
        rolling_calendar_metrics(x)
    )


# ============================================================
# 9. SYNCHRONIZED CP MONITORING LAYER
#
# Forward fill permitted ONLY HERE.
# Max = five business days.
# ============================================================

cp_raw2 = pd.concat({

    "CP_Credit_30D": CP2["30D"],
    "CP_Credit_60D": CP2["60D"],
    "CP_Credit_90D": CP2["90D"]

}, axis=1).sort_index()

calendar = pd.date_range(
    cp_raw2.index.min(),
    cp_raw2.index.max(),
    freq="B"
)

cp_monitor2 = (
    cp_raw2
    .reindex(calendar)
    .ffill(limit=5)
)

cp_monitor2.index.name = "Date"


# ============================================================
# 10. CORRECTED HISTORICAL CP COMPOSITE
#
# Genuine trailing-calendar-year z-score at each date.
#
# We calculate z from the MONITORING series because the
# composite itself is explicitly the synchronized monitor.
# ============================================================

cp_z2 = pd.DataFrame(
    index=cp_monitor2.index
)

for maturity in ["30D", "60D", "90D"]:

    col = f"CP_Credit_{maturity}"

    x = cp_monitor2[col]

    zvals = []

    for dt in x.index:

        hist = x.loc[:dt].dropna()

        w = hist.loc[
            hist.index >= dt - pd.DateOffset(years=1)
        ]

        # Require enough information for a meaningful z-score.
        if len(w) < 20:
            zvals.append(np.nan)
        else:
            zvals.append(
                zscore_current(w)
            )

    cp_z2[maturity] = zvals


valid_n = cp_z2.notna().sum(axis=1)

cp_monitor2["CP_Stress_Composite"] = (
    cp_z2.mean(
        axis=1,
        skipna=True
    )
)

cp_monitor2.loc[
    valid_n < 2,
    "CP_Stress_Composite"
] = np.nan

cp_monitor2["CP_Regime"] = (
    cp_monitor2[
        "CP_Stress_Composite"
    ].apply(cp_regime)
)


# ============================================================
# 11. ADD RAW-ESTIMATION ROLLING METRICS TO MONITOR FILE
#
# Carrying them onto the monitoring calendar does NOT alter
# their estimation; calculations above used raw observations.
# ============================================================

for maturity in ["30D", "60D", "90D"]:

    a = cp_analysis2[maturity]

    for metric in [
        "Z_1Y",
        "Z_3Y",
        "Pct_1Y",
        "Pct_3Y",
        "AR1_1Y",
        "HalfLife_1Y_obs",
        "AR1_3Y",
        "HalfLife_3Y_obs",
        "Obs_1Y",
        "Obs_3Y"
    ]:

        cp_monitor2[
            f"CP_{maturity}_{metric}"
        ] = (
            a[metric]
            .reindex(cp_monitor2.index)
            .ffill(limit=5)
        )


# ============================================================
# 12. TERM STRUCTURE
# ============================================================

cp_monitor2["CP_Credit_90D_minus_30D"] = (
    cp_monitor2["CP_Credit_90D"]
    - cp_monitor2["CP_Credit_30D"]
)

cp_monitor2["CP_Credit_60D_minus_30D"] = (
    cp_monitor2["CP_Credit_60D"]
    - cp_monitor2["CP_Credit_30D"]
)


# ============================================================
# 13. MASTER HISTORY
# ============================================================

master2 = pd.concat(
    [
        rates2.add_prefix("RATE_"),
        spreads2.add_prefix("SPREAD_"),
        cp_monitor2
    ],
    axis=1
).sort_index()

master2.index.name = "Date"


# ============================================================
# 14. WINDOW VALIDATION
#
# Explicitly demonstrate that "1Y" now means calendar year.
# ============================================================

window_audit = []

for maturity, x in CP2.items():

    x = x.dropna()

    end = x.index[-1]

    x1 = calendar_window(x, 1)
    x3 = calendar_window(x, 3)

    window_audit.append({

        "Series":
            f"CP {maturity}",

        "Latest":
            end.date(),

        "1Y_Cutoff":
            (end - pd.DateOffset(years=1)).date(),

        "1Y_First_Actual":
            x1.index[0].date(),

        "1Y_Obs":
            len(x1),

        "3Y_Cutoff":
            (end - pd.DateOffset(years=3)).date(),

        "3Y_First_Actual":
            x3.index[0].date(),

        "3Y_Obs":
            len(x3)
    })

window_audit = pd.DataFrame(
    window_audit
)


# ============================================================
# 15. CURRENT COMPOSITE
# ============================================================

comp = (
    cp_monitor2["CP_Stress_Composite"]
    .dropna()
)

latest_comp_date = comp.index[-1]
latest_comp = comp.iloc[-1]


# ============================================================
# 16. EXPORT AUDIT VERSION ONLY
# ============================================================

rates2.to_csv(
    f"{OUT2}/raw_rates.csv"
)

spreads2.to_csv(
    f"{OUT2}/spreads.csv"
)

analytics2.to_csv(
    f"{OUT2}/latest_spreads.csv",
    index=False
)

cp_snapshot2.to_csv(
    f"{OUT2}/latest_cp.csv",
    index=False
)

cp_monitor2.to_csv(
    f"{OUT2}/cp_monitor.csv"
)

master2.to_csv(
    f"{OUT2}/master_history.csv"
)

master2.to_parquet(
    f"{OUT2}/master_history.parquet"
)

spread_audit.to_csv(
    f"{OUT2}/spread_arithmetic_audit.csv",
    index=False
)

window_audit.to_csv(
    f"{OUT2}/window_audit.csv",
    index=False
)


# ============================================================
# 17. CORRECTED METADATA
# ============================================================

metadata2 = {

    "project":
        "U.S. Funding Conditions",

    "version":
        "2.0-audit",

    "units": {
        "raw_rates":
            "percent",
        "spreads":
            "basis points",
        "z_scores":
            "standard deviations",
        "half_life":
            "observations"
    },

    "methodology": {

        "calendar_windows":
            "Metrics labelled 1Y use observations occurring "
            "within the trailing one calendar year. Metrics "
            "labelled 3Y use observations occurring within the "
            "trailing three calendar years. Observation-count "
            "proxies such as 252/756 are not used.",

        "spread_history":
            "Each spread uses dates on which both constituent "
            "raw series have observations. Statistical spread "
            "series are not forward-filled.",

        "cp_composite":
            "Mean trailing-one-calendar-year standardized "
            "A2/P2 minus AA nonfinancial CP spread across "
            "30D, 60D and 90D maturities. At least two "
            "maturities are required.",

        "monitor_alignment":
            "The synchronized CP monitoring layer permits a "
            "maximum five-business-day forward fill. This "
            "alignment is not used when estimating raw-series "
            "AR(1), z-score, percentile or ADF statistics.",

        "ar1":
            "x_t = c + phi*x_(t-1) + error.",

        "half_life":
            "AR(1)-implied mean-reversion half-life = "
            "-ln(2)/ln(phi), reported in observations and only "
            "where 0 < phi < 1.",

        "adf":
            "Augmented Dickey-Fuller test with constant and "
            "AIC lag selection on the full raw spread history."
    }
}

with open(
    f"{OUT2}/metadata.json",
    "w"
) as f:

    json.dump(
        metadata2,
        f,
        indent=4
    )


# ============================================================
# 18. OUTPUT — AUDIT BEFORE WE TOUCH GITHUB
# ============================================================

print(
    "\n========== SPREAD ARITHMETIC AUDIT ==========\n"
)

display(spread_audit)

print(
    "\nALL SPREADS PASS:",
    bool(spread_audit["PASS"].all())
)


print(
    "\n========== CALENDAR-WINDOW AUDIT ==========\n"
)

display(window_audit)


print(
    "\n========== CORRECTED CP SNAPSHOT ==========\n"
)

display(
    cp_snapshot2.round(3)
)


print(
    "\n========== CORRECTED SPREAD SNAPSHOT ==========\n"
)

display(
    analytics2[
        [
            "Spread",
            "End",
            "Current_bp",
            "Full_Percentile",
            "Pct_1Y",
            "Z_Full",
            "Z_1Y",
            "Obs_1Y",
            "AR1_1Y",
            "HalfLife_1Y_obs",
            "Z_3Y",
            "Obs_3Y",
            "AR1_3Y",
            "HalfLife_3Y_obs",
            "ADF_p",
            "Regime"
        ]
    ].round(3)
)


print(
    "\n========== CORRECTED CP COMPOSITE ==========\n"
)

print(
    "Date:",
    latest_comp_date.date()
)

print(
    "Composite:",
    round(latest_comp, 3)
)

print(
    "Regime:",
    cp_regime(latest_comp)
)


print(
    "\n========== V2 AUDIT FILES ==========\n"
)

for f in sorted(os.listdir(OUT2)):

    print(f)


print(
    "\nNO V1 FILES OR GITHUB FILES HAVE BEEN OVERWRITTEN."
)
```

```python
import os
import re
import subprocess
from getpass import getpass

REPO = "/content/U.S.-Funding-Conditions"
APP = os.path.join(REPO, "app.py")

# ============================================================
# LOAD CURRENT PRODUCTION APP
# ============================================================

with open(APP, "r", encoding="utf-8") as f:
    code = f.read()

original = code


# ============================================================
# 1. ADD make_subplots IMPORT
# ============================================================

old_import = "import plotly.graph_objects as go\n"

new_import = (
    "import plotly.graph_objects as go\n"
    "from plotly.subplots import make_subplots\n"
)

if "from plotly.subplots import make_subplots" not in code:

    if old_import not in code:
        raise RuntimeError("Plotly import anchor not found")

    code = code.replace(
        old_import,
        new_import,
        1
    )


# ============================================================
# 2. REPLACE ONLY line_chart()
#
# Nothing below touches:
# - data loading
# - CSV/parquet files
# - analytics
# - AR(1)
# - half-life
# - z-scores
# - regimes
# - calendar-window controls
# ============================================================

start = code.find("def line_chart(")
end = code.find("\ndef chart_date_control(", start)

if start == -1 or end == -1:
    raise RuntimeError(
        "Could not isolate current line_chart() safely"
    )

old_line_chart = code[start:end]


new_line_chart = r'''def line_chart(
    df,
    columns,
    title,
    ytitle,
    zero_line=False
):
    """
    Presentation-only chart renderer.

    Standard:
        Original Plotly chart with full linear y-axis.

    Outlier-aware:
        Core observations remain on a readable linear scale.
        Extreme observations are NOT deleted, winsorized,
        clipped or used to recalculate anything. They are
        displayed at their true values in a separate upper
        panel.

    The input dataframe is never modified.
    """

    available = [
        col for col in columns
        if col in df.columns
    ]

    # --------------------------------------------------------
    # DISPLAY MODE
    # --------------------------------------------------------

    mode = st.radio(
        "Chart display",
        [
            "Outlier-aware",
            "Standard"
        ],
        horizontal=True,
        key=f"display_mode_{title}"
    )

    # --------------------------------------------------------
    # STANDARD VIEW
    # --------------------------------------------------------

    if mode == "Standard":

        fig = go.Figure()

        for col in available:

            x = df[col].dropna()

            fig.add_trace(
                go.Scatter(
                    x=x.index,
                    y=x.values,
                    mode="lines",
                    name=col,
                    line=dict(width=1.6)
                )
            )

        if zero_line:
            fig.add_hline(
                y=0,
                line_dash="dot",
                opacity=0.5
            )

        fig.update_layout(
            title=title,
            yaxis_title=ytitle,
            xaxis_title="",
            hovermode="x unified",
            height=480,
            margin=dict(
                l=20,
                r=20,
                t=55,
                b=20
            ),
            legend=dict(
                orientation="h",
                yanchor="bottom",
                y=1.02,
                xanchor="left",
                x=0
            )
        )

        return fig

    # --------------------------------------------------------
    # DETERMINE WHETHER OUTLIER PANEL IS ACTUALLY NEEDED
    #
    # Thresholds are calculated ONLY from the dataframe
    # already supplied for display (e.g. selected 3Y period).
    #
    # This does not modify source data or analytics.
    # --------------------------------------------------------

    arrays = []

    for col in available:

        values = pd.to_numeric(
            df[col],
            errors="coerce"
        )

        values = values[
            np.isfinite(values)
        ]

        if len(values):
            arrays.append(
                values.to_numpy()
            )

    if not arrays:

        return go.Figure()

    all_values = np.concatenate(arrays)

    # Too little data -> ordinary chart.
    if len(all_values) < 20:

        fig = go.Figure()

        for col in available:

            x = df[col].dropna()

            fig.add_trace(
                go.Scatter(
                    x=x.index,
                    y=x.values,
                    mode="lines",
                    name=col,
                    line=dict(width=1.6)
                )
            )

        if zero_line:
            fig.add_hline(
                y=0,
                line_dash="dot",
                opacity=0.5
            )

        fig.update_layout(
            title=title,
            yaxis_title=ytitle,
            xaxis_title="",
            hovermode="x unified",
            height=480,
            margin=dict(
                l=20,
                r=20,
                t=55,
                b=20
            )
        )

        return fig

    # --------------------------------------------------------
    # CORE RANGE
    #
    # 1st / 99th percentiles identify observations that
    # dominate the linear scale.
    #
    # IMPORTANT:
    # They are used only to decide which PANEL displays an
    # observation. Values themselves are never changed.
    # --------------------------------------------------------

    lower = float(
        np.nanpercentile(
            all_values,
            1
        )
    )

    upper = float(
        np.nanpercentile(
            all_values,
            99
        )
    )

    if (
        not np.isfinite(lower)
        or not np.isfinite(upper)
        or upper <= lower
    ):

        lower = float(
            np.nanmin(all_values)
        )

        upper = float(
            np.nanmax(all_values)
        )

    extreme_mask_all = (
        (all_values < lower)
        | (all_values > upper)
    )

    has_extremes = bool(
        np.any(extreme_mask_all)
    )

    # If there is effectively nothing to separate,
    # return ordinary full-range chart.
    if not has_extremes:

        fig = go.Figure()

        for col in available:

            x = df[col].dropna()

            fig.add_trace(
                go.Scatter(
                    x=x.index,
                    y=x.values,
                    mode="lines",
                    name=col,
                    line=dict(width=1.6)
                )
            )

        if zero_line:
            fig.add_hline(
                y=0,
                line_dash="dot",
                opacity=0.5
            )

        fig.update_layout(
            title=title,
            yaxis_title=ytitle,
            xaxis_title="",
            hovermode="x unified",
            height=480,
            margin=dict(
                l=20,
                r=20,
                t=55,
                b=20
            )
        )

        return fig

    # --------------------------------------------------------
    # TWO-PANEL OUTLIER-AWARE VIEW
    #
    # TOP:
    # extreme observations at TRUE values
    #
    # BOTTOM:
    # core observations at TRUE values
    #
    # No clipping / winsorisation / transformation.
    # --------------------------------------------------------

    fig = make_subplots(
        rows=2,
        cols=1,
        shared_xaxes=True,
        vertical_spacing=0.06,
        row_heights=[
            0.27,
            0.73
        ]
    )

    for col in available:

        series = pd.to_numeric(
            df[col],
            errors="coerce"
        )

        valid = series.dropna()

        extreme = valid[
            (valid < lower)
            | (valid > upper)
        ]

        core = valid[
            (valid >= lower)
            & (valid <= upper)
        ]

        # -----------------------------
        # EXTREMES: markers only.
        #
        # We deliberately do not join
        # isolated extremes with lines,
        # because that would imply a
        # continuous path across missing
        # core observations.
        # -----------------------------

        if len(extreme):

            fig.add_trace(
                go.Scatter(
                    x=extreme.index,
                    y=extreme.values,
                    mode="markers",
                    name=f"{col} — extremes",
                    legendgroup=col,
                    marker=dict(
                        size=7,
                        symbol="diamond"
                    ),
                    customdata=np.array(
                        [
                            ["Extreme"]
                            for _ in range(
                                len(extreme)
                            )
                        ]
                    ),
                    hovertemplate=(
                        "%{x}<br>"
                        + col
                        + ": %{y:.4g}<br>"
                        "Extreme observation"
                        "<extra></extra>"
                    )
                ),
                row=1,
                col=1
            )

        # -----------------------------
        # CORE
        # -----------------------------

        if len(core):

            fig.add_trace(
                go.Scatter(
                    x=core.index,
                    y=core.values,
                    mode="lines",
                    name=col,
                    legendgroup=col,
                    line=dict(
                        width=1.6
                    ),
                    hovertemplate=(
                        "%{x}<br>"
                        + col
                        + ": %{y:.4g}"
                        "<extra></extra>"
                    )
                ),
                row=2,
                col=1
            )

        # -----------------------------
        # BOUNDARY INDICATORS
        #
        # Show where observations leave
        # the core panel. These markers
        # sit at the panel boundary but
        # hover displays the TRUE value.
        # -----------------------------

        if len(extreme):

            boundary_y = np.where(
                extreme.values > upper,
                upper,
                lower
            )

            symbols = np.where(
                extreme.values > upper,
                "triangle-up",
                "triangle-down"
            )

            fig.add_trace(
                go.Scatter(
                    x=extreme.index,
                    y=boundary_y,
                    mode="markers",
                    name=f"{col} — boundary",
                    legendgroup=col,
                    showlegend=False,
                    marker=dict(
                        size=7,
                        symbol=symbols
                    ),
                    customdata=extreme.values,
                    hovertemplate=(
                        "%{x}<br>"
                        + col
                        + " true value: "
                        "%{customdata:.4g}"
                        "<br>Shown in extremes panel"
                        "<extra></extra>"
                    )
                ),
                row=2,
                col=1
            )

    # --------------------------------------------------------
    # ZERO LINE — CORE PANEL ONLY
    # --------------------------------------------------------

    if zero_line and lower <= 0 <= upper:

        fig.add_hline(
            y=0,
            line_dash="dot",
            opacity=0.5,
            row=2,
            col=1
        )

    # --------------------------------------------------------
    # CORE PANEL RANGE
    #
    # Small padding only.
    # This is NOT clipping observations because observations
    # outside the range are displayed at their true values
    # in the upper panel.
    # --------------------------------------------------------

    core_span = upper - lower

    if core_span > 0:

        pad = core_span * 0.05

        fig.update_yaxes(
            range=[
                lower - pad,
                upper + pad
            ],
            row=2,
            col=1
        )

    # --------------------------------------------------------
    # LABELS
    # --------------------------------------------------------

    fig.update_yaxes(
        title_text="Extremes",
        row=1,
        col=1
    )

    fig.update_yaxes(
        title_text=ytitle,
        row=2,
        col=1
    )

    fig.update_xaxes(
        title_text="",
        row=2,
        col=1
    )

    fig.update_layout(
        title=title,
        hovermode="x",
        height=600,
        margin=dict(
            l=20,
            r=20,
            t=55,
            b=20
        ),
        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=1.02,
            xanchor="left",
            x=0
        )
    )

    return fig


'''

code = (
    code[:start]
    + new_line_chart
    + code[end:]
)


# ============================================================
# 3. WRITE APP.PY ONLY
# ============================================================

if code == original:
    raise RuntimeError(
        "No app.py change generated"
    )

with open(APP, "w", encoding="utf-8") as f:
    f.write(code)


# ============================================================
# 4. PYTHON SYNTAX VALIDATION
# ============================================================

subprocess.run(
    [
        "python",
        "-m",
        "py_compile",
        "app.py"
    ],
    cwd=REPO,
    check=True
)

print("Syntax: PASS")


# ============================================================
# 5. STRUCTURAL VALIDATION
# ============================================================

with open(APP, "r", encoding="utf-8") as f:
    check = f.read()

assert (
    "from plotly.subplots import make_subplots"
    in check
)

assert (
    '"Outlier-aware"'
    in check
)

assert (
    '"Standard"'
    in check
)

assert (
    "Extreme observation"
    in check
)

assert (
    "def chart_date_control("
    in check
)

print("Outlier-aware chart helper: PASS")
print("Calendar chart controls: PASS")


# ============================================================
# 6. CRITICAL SAFETY CHECK:
#    ONLY app.py MAY BE STAGED
# ============================================================

status = subprocess.run(
    [
        "git",
        "status",
        "--porcelain"
    ],
    cwd=REPO,
    capture_output=True,
    text=True,
    check=True
).stdout

print("\n========== GIT STATUS ==========")
print(status or "Clean")

changed = []

for line in status.splitlines():

    if not line.strip():
        continue

    path = line[3:].strip()

    # handle rename syntax defensively
    if " -> " in path:
        path = path.split(" -> ")[-1]

    changed.append(path)

unexpected = [
    p for p in changed
    if p != "app.py"
]

if unexpected:

    raise RuntimeError(
        "STOP: files other than app.py are modified: "
        + ", ".join(unexpected)
    )

print(
    "Safety check: ONLY app.py modified"
)


# ============================================================
# 7. SHOW DIFF SUMMARY BEFORE COMMIT
# ============================================================

print("\n========== DIFF SUMMARY ==========\n")

subprocess.run(
    [
        "git",
        "diff",
        "--stat",
        "--",
        "app.py"
    ],
    cwd=REPO,
    check=True
)


# ============================================================
# 8. COMMIT ONLY APP.PY
# ============================================================

subprocess.run(
    [
        "git",
        "add",
        "--",
        "app.py"
    ],
    cwd=REPO,
    check=True
)

staged = subprocess.run(
    [
        "git",
        "diff",
        "--cached",
        "--name-only"
    ],
    cwd=REPO,
    capture_output=True,
    text=True,
    check=True
).stdout.splitlines()

if staged != ["app.py"]:

    raise RuntimeError(
        "STOP: staged files are not exactly ['app.py']: "
        + repr(staged)
    )

print(
    "Staging safety check: PASS"
)

subprocess.run(
    [
        "git",
        "commit",
        "-m",
        "Add non-destructive outlier-aware chart display"
    ],
    cwd=REPO,
    check=True
)


# ============================================================
# 9. PUSH
# ============================================================

token = getpass(
    "\nPaste GitHub PAT: "
)

auth_url = (
    f"https://YieldShock13:{token}"
    "@github.com/YieldShock13/"
    "U.S.-Funding-Conditions.git"
)

subprocess.run(
    [
        "git",
        "remote",
        "set-url",
        "origin",
        auth_url
    ],
    cwd=REPO,
    check=True
)

try:

    subprocess.run(
        [
            "git",
            "push",
            "origin",
            "main"
        ],
        cwd=REPO,
        check=True
    )

finally:

    subprocess.run(
        [
            "git",
            "remote",
            "set-url",
            "origin",
            "https://github.com/"
            "YieldShock13/"
            "U.S.-Funding-Conditions.git"
        ],
        cwd=REPO,
        check=True
    )

    token = None


print("\n========================================")
print("OUTLIER-AWARE DISPLAY DEPLOYED")
print("========================================")
print("Modified file: app.py ONLY")
print("Data files: UNCHANGED")
print("Data pipeline: UNCHANGED")
print("Analytics: UNCHANGED")
print("Calendar-window methodology: UNCHANGED")
print("Extreme values: RETAINED AT TRUE VALUES")
print("Standard full-range view: STILL AVAILABLE")
```
