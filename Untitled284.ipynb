```python
# ============================================================
# SA RATES RV / PERSISTENCE MONITOR
# Automated RBond pull
# ============================================================

import io
import requests
import numpy as np
import pandas as pd
import plotly.graph_objects as go
from IPython.display import display
from statsmodels.api import OLS, add_constant

# ------------------------------------------------------------
# 1. DOWNLOAD
# ------------------------------------------------------------

URL = "https://rbond.co.za/api/v1/history.csv"

r = requests.get(URL, timeout=90)
r.raise_for_status()

raw = pd.read_csv(io.StringIO(r.text))

raw["date"] = pd.to_datetime(raw["date"])
raw["yield_pct"] = pd.to_numeric(raw["yield_pct"], errors="coerce")

wanted = ["ZARONIA", "ZARGB1", "ZARGB5", "ZARGB10"]

raw = raw[
    raw["bond"].isin(wanted)
].copy()

# ZARONIA only actually available from Sep-2023,
# so no reason to retain earlier observations.
raw = raw[
    raw["date"] >= "2023-09-25"
]

rates = (
    raw.pivot_table(
        index="date",
        columns="bond",
        values="yield_pct",
        aggfunc="last"
    )
    .sort_index()
)

print("=" * 75)
print("UNDERLYING DATA")
print("=" * 75)

availability = pd.DataFrame({
    "First": rates.apply(lambda x: x.dropna().index.min()),
    "Last": rates.apply(lambda x: x.dropna().index.max()),
    "Observations": rates.count()
})

display(availability)


# ------------------------------------------------------------
# 2. CONSTRUCT SPREADS
# All expressed in basis points
# ------------------------------------------------------------

spreads = pd.DataFrame(index=rates.index)

spreads["1Y - ZARONIA"] = (
    rates["ZARGB1"] - rates["ZARONIA"]
) * 100

spreads["5Y - ZARONIA"] = (
    rates["ZARGB5"] - rates["ZARONIA"]
) * 100

spreads["10Y - ZARONIA"] = (
    rates["ZARGB10"] - rates["ZARONIA"]
) * 100

spreads["5Y - 1Y"] = (
    rates["ZARGB5"] - rates["ZARGB1"]
) * 100

spreads["10Y - 1Y"] = (
    rates["ZARGB10"] - rates["ZARGB1"]
) * 100

spreads["10Y - 5Y"] = (
    rates["ZARGB10"] - rates["ZARGB5"]
) * 100


# ------------------------------------------------------------
# 3. AR(1)
#
# S_t = alpha + phi*S_(t-1) + epsilon_t
#
# Half-life = -ln(2) / ln(phi)
# only reported where 0 < phi < 1
# ------------------------------------------------------------

def ar1(series):

    s = series.dropna().astype(float)

    lagged = pd.DataFrame({
        "y": s,
        "lag": s.shift(1)
    }).dropna()

    if len(lagged) < 30:
        return np.nan, np.nan, np.nan

    model = OLS(
        lagged["y"],
        add_constant(lagged["lag"])
    ).fit()

    phi = float(model.params["lag"])
    r2 = float(model.rsquared)

    if 0 < phi < 1:
        half_life = -np.log(2) / np.log(phi)
    else:
        half_life = np.nan

    return phi, half_life, r2


# ------------------------------------------------------------
# 4. SUMMARY
# ------------------------------------------------------------

results = []

for name in spreads.columns:

    s = spreads[name].dropna()

    current = s.iloc[-1]
    current_date = s.index[-1]

    # 60 observations, not calendar days
    if len(s) >= 61:
        change_60 = current - s.iloc[-61]
    else:
        change_60 = np.nan

    # Rolling 252-observation z-score
    trailing = s.tail(252)

    if len(trailing) >= 60 and trailing.std(ddof=1) != 0:
        z = (
            current - trailing.mean()
        ) / trailing.std(ddof=1)
    else:
        z = np.nan

    # Full-history percentile
    percentile = (
        (s <= current).sum() / len(s)
    ) * 100

    phi, half_life, r2 = ar1(s)

    results.append({
        "Spread": name,
        "As Of": current_date.date(),
        "Current (bp)": current,
        "60D Δ (bp)": change_60,
        "252D Z": z,
        "Historical %ile": percentile,
        "AR(1) φ": phi,
        "Half-Life (days)": half_life,
        "AR(1) R²": r2,
        "N": len(s)
    })

summary = (
    pd.DataFrame(results)
    .set_index("Spread")
)

print("\n" + "=" * 100)
print("SA RATES RV / PERSISTENCE MONITOR")
print("=" * 100)

display(
    summary.style.format({
        "Current (bp)": "{:.2f}",
        "60D Δ (bp)": "{:+.2f}",
        "252D Z": "{:+.2f}",
        "Historical %ile": "{:.1f}",
        "AR(1) φ": "{:.4f}",
        "Half-Life (days)": "{:.1f}",
        "AR(1) R²": "{:.3f}",
        "N": "{:.0f}"
    })
)


# ------------------------------------------------------------
# 5. ROLLING 252-OBSERVATION Z-SCORES
# ------------------------------------------------------------

zrolling = pd.DataFrame(index=spreads.index)

for name in spreads.columns:

    mean = spreads[name].rolling(
        252,
        min_periods=60
    ).mean()

    std = spreads[name].rolling(
        252,
        min_periods=60
    ).std()

    zrolling[name] = (
        spreads[name] - mean
    ) / std


# ------------------------------------------------------------
# 6. CHART FUNCTION
# ------------------------------------------------------------

def plot_lines(df, columns, title, ytitle):

    fig = go.Figure()

    for col in columns:

        x = df[col].dropna()

        fig.add_trace(
            go.Scatter(
                x=x.index,
                y=x.values,
                mode="lines",
                name=col,
                line=dict(width=1.7)
            )
        )

    fig.update_layout(
        title=title,
        xaxis_title="Date",
        yaxis_title=ytitle,
        hovermode="x unified",
        template="plotly_white",
        height=600,
        legend=dict(
            orientation="h",
            y=1.08
        )
    )

    fig.show()


# ------------------------------------------------------------
# 7. CHART — ZARONIA SPREADS
# ------------------------------------------------------------

plot_lines(
    spreads,
    [
        "1Y - ZARONIA",
        "5Y - ZARONIA",
        "10Y - ZARONIA"
    ],
    "SAGB Constant-Maturity Spreads vs ZARONIA",
    "Basis Points"
)


# ------------------------------------------------------------
# 8. CHART — CURVE SPREADS
# ------------------------------------------------------------

plot_lines(
    spreads,
    [
        "5Y - 1Y",
        "10Y - 1Y",
        "10Y - 5Y"
    ],
    "South African Rates Curve Spreads",
    "Basis Points"
)


# ------------------------------------------------------------
# 9. CHART — ROLLING Z-SCORES
# ------------------------------------------------------------

fig = go.Figure()

for col in zrolling.columns:

    x = zrolling[col].dropna()

    fig.add_trace(
        go.Scatter(
            x=x.index,
            y=x.values,
            mode="lines",
            name=col,
            line=dict(width=1.5)
        )
    )

fig.add_hline(y=0, line_dash="dash")
fig.add_hline(y=2, line_dash="dot")
fig.add_hline(y=-2, line_dash="dot")

fig.update_layout(
    title="SA Rates RV — Rolling 252-Observation Z-Scores",
    xaxis_title="Date",
    yaxis_title="Z-Score",
    hovermode="x unified",
    template="plotly_white",
    height=650,
    legend=dict(
        orientation="h",
        y=1.10
    )
)

fig.show()


# ------------------------------------------------------------
# 10. SAVE DASHBOARD DATA
# ------------------------------------------------------------

dashboard = pd.concat(
    [
        rates.add_suffix("_RATE"),
        spreads.add_suffix("_SPREAD_BP"),
        zrolling.add_suffix("_ZSCORE")
    ],
    axis=1
)

dashboard.to_csv(
    "SA_Rates_RV_Dashboard.csv"
)

summary.to_csv(
    "SA_Rates_RV_Summary.csv"
)

print("\nSaved:")
print("SA_Rates_RV_Dashboard.csv")
print("SA_Rates_RV_Summary.csv")

# ============================================================
# ROLLING PERSISTENCE
# 252-observation rolling AR(1)
# ============================================================

st.subheader("Rolling Persistence")

ROLLING_AR_WINDOW = 252
MIN_AR_OBS = 120


def rolling_ar1(series, window=252, min_obs=120):

    s = series.dropna().astype(float)

    phi_series = pd.Series(
        index=s.index,
        dtype=float
    )

    half_life_series = pd.Series(
        index=s.index,
        dtype=float
    )

    for i in range(len(s)):

        start = max(
            0,
            i - window + 1
        )

        sample = s.iloc[start:i + 1]

        if len(sample) < min_obs:
            continue

        reg = pd.DataFrame({
            "y": sample,
            "lag": sample.shift(1)
        }).dropna()

        X = sm.add_constant(
            reg["lag"]
        )

        model = sm.OLS(
            reg["y"],
            X
        ).fit()

        phi = float(
            model.params["lag"]
        )

        phi_series.iloc[i] = phi

        if 0 < phi < 1:

            half_life_series.iloc[i] = (
                -np.log(2)
                / np.log(phi)
            )

    return phi_series, half_life_series


rolling_phi, rolling_hl = rolling_ar1(
    series,
    window=ROLLING_AR_WINDOW,
    min_obs=MIN_AR_OBS
)


# ------------------------------------------------------------
# ROLLING PHI CHART
# ------------------------------------------------------------

fig_phi = go.Figure()

fig_phi.add_trace(
    go.Scatter(
        x=rolling_phi.index,
        y=rolling_phi.values,
        mode="lines",
        name="Rolling AR(1) φ",
        line=dict(width=2)
    )
)

fig_phi.add_hline(
    y=1,
    line_dash="dash",
    annotation_text="φ = 1"
)

fig_phi.update_layout(
    title=f"{selected} — Rolling AR(1) Persistence",
    xaxis_title="Date",
    yaxis_title="AR(1) φ",
    hovermode="x unified",
    template="plotly_white",
    height=450,
    margin=dict(
        l=20,
        r=20,
        t=50,
        b=20
    )
)

st.plotly_chart(
    fig_phi,
    use_container_width=True
)


# ------------------------------------------------------------
# ROLLING HALF-LIFE CHART
# ------------------------------------------------------------

fig_hl = go.Figure()

fig_hl.add_trace(
    go.Scatter(
        x=rolling_hl.index,
        y=rolling_hl.values,
        mode="lines",
        name="Rolling Half-Life",
        line=dict(width=2)
    )
)

fig_hl.update_layout(
    title=f"{selected} — Rolling Implied Half-Life",
    xaxis_title="Date",
    yaxis_title="Trading Days",
    hovermode="x unified",
    template="plotly_white",
    height=450,
    margin=dict(
        l=20,
        r=20,
        t=50,
        b=20
    )
)

st.plotly_chart(
    fig_hl,
    use_container_width=True
)
```

```python
# ============================================================
# SA RATES — ROLLING PERSISTENCE
# COLAB ONLY
#
# Requires existing dataframe: spreads
#
# Outputs:
#   1. Rolling 252-observation AR(1) phi
#   2. Rolling implied half-life
#   3. Latest rolling persistence table
# ============================================================

import numpy as np
import pandas as pd
import statsmodels.api as sm
import plotly.graph_objects as go
from IPython.display import display


# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

WINDOW = 252
MIN_OBS = 120


# ------------------------------------------------------------
# ROLLING AR(1)
#
# S_t = alpha + phi*S_(t-1) + epsilon_t
#
# Half-life only defined here where:
# 0 < phi < 1
# ------------------------------------------------------------

def rolling_ar1(series, window=252, min_obs=120):

    s = series.dropna().astype(float)

    phi_out = pd.Series(
        np.nan,
        index=s.index,
        dtype=float
    )

    hl_out = pd.Series(
        np.nan,
        index=s.index,
        dtype=float
    )

    for i in range(len(s)):

        start = max(0, i - window + 1)

        sample = s.iloc[start:i + 1]

        if len(sample) < min_obs:
            continue

        reg = pd.DataFrame({
            "y": sample,
            "lag": sample.shift(1)
        }).dropna()

        if len(reg) < min_obs - 1:
            continue

        X = sm.add_constant(
            reg["lag"]
        )

        model = sm.OLS(
            reg["y"],
            X
        ).fit()

        phi = float(
            model.params["lag"]
        )

        phi_out.iloc[i] = phi

        if 0 < phi < 1:

            hl_out.iloc[i] = (
                -np.log(2)
                / np.log(phi)
            )

    return phi_out, hl_out


# ------------------------------------------------------------
# CALCULATE ALL SIX
# ------------------------------------------------------------

rolling_phi = pd.DataFrame(
    index=spreads.index
)

rolling_half_life = pd.DataFrame(
    index=spreads.index
)


for col in spreads.columns:

    phi, hl = rolling_ar1(
        spreads[col],
        window=WINDOW,
        min_obs=MIN_OBS
    )

    rolling_phi[col] = phi.reindex(
        spreads.index
    )

    rolling_half_life[col] = hl.reindex(
        spreads.index
    )


# ------------------------------------------------------------
# LATEST RESULTS
# ------------------------------------------------------------

latest_rows = []

for col in spreads.columns:

    p = rolling_phi[col].dropna()
    h = rolling_half_life[col].dropna()

    latest_phi = (
        p.iloc[-1]
        if len(p)
        else np.nan
    )

    # IMPORTANT:
    # Only report half-life for the SAME latest phi observation.
    latest_date = (
        p.index[-1]
        if len(p)
        else None
    )

    if latest_date is not None:

        latest_hl = rolling_half_life.loc[
            latest_date,
            col
        ]

    else:
        latest_hl = np.nan

    latest_rows.append({
        "Spread": col,
        "As Of": latest_date,
        "Rolling AR(1) φ": latest_phi,
        "Implied Half-Life (days)": latest_hl
    })


latest_persistence = (
    pd.DataFrame(latest_rows)
    .set_index("Spread")
)


print("=" * 85)
print("LATEST 252-OBSERVATION ROLLING PERSISTENCE")
print("=" * 85)

display(
    latest_persistence.style.format({
        "Rolling AR(1) φ": "{:.4f}",
        "Implied Half-Life (days)": "{:.1f}"
    })
)


# ============================================================
# CHART 1 — ROLLING AR(1) PHI
# ============================================================

fig = go.Figure()

for col in rolling_phi.columns:

    x = rolling_phi[col].dropna()

    fig.add_trace(
        go.Scatter(
            x=x.index,
            y=x.values,
            mode="lines",
            name=col,
            line=dict(width=1.7)
        )
    )


fig.add_hline(
    y=1.0,
    line_dash="dash",
    annotation_text="φ = 1"
)


fig.update_layout(
    title="SA Rates Spreads — Rolling 252-Observation AR(1) Persistence",
    xaxis_title="Date",
    yaxis_title="AR(1) φ",
    hovermode="x unified",
    template="plotly_white",
    height=600,
    legend=dict(
        orientation="h",
        y=1.13
    )
)

fig.show()


# ============================================================
# CHART 2 — ROLLING IMPLIED HALF-LIFE
# ============================================================

fig = go.Figure()

for col in rolling_half_life.columns:

    x = rolling_half_life[col].dropna()

    fig.add_trace(
        go.Scatter(
            x=x.index,
            y=x.values,
            mode="lines",
            name=col,
            line=dict(width=1.7)
        )
    )


fig.update_layout(
    title="SA Rates Spreads — Rolling AR(1) Implied Half-Life",
    xaxis_title="Date",
    yaxis_title="Trading Days",
    hovermode="x unified",
    template="plotly_white",
    height=600,
    legend=dict(
        orientation="h",
        y=1.13
    )
)

fig.show()
```

```python
# ============================================================
# SA RATES — ROLLING AR(1) PERSISTENCE
# FINAL COLAB VERSION
#
# Chart: rolling AR(1) phi
# Table: current phi + half-life + persistence flag
# ============================================================

import numpy as np
import pandas as pd
import statsmodels.api as sm
import plotly.graph_objects as go
from IPython.display import display

WINDOW = 252
MIN_OBS = 120

# Above this, half-life becomes numerically unstable / unhelpful
NEAR_UNIT_ROOT = 0.995


# ------------------------------------------------------------
# ROLLING AR(1)
# ------------------------------------------------------------

def rolling_ar1_phi(series, window=252, min_obs=120):

    s = series.dropna().astype(float)

    output = pd.Series(
        np.nan,
        index=s.index,
        dtype=float
    )

    for i in range(len(s)):

        sample = s.iloc[
            max(0, i-window+1):i+1
        ]

        if len(sample) < min_obs:
            continue

        reg = pd.DataFrame({
            "y": sample,
            "lag": sample.shift(1)
        }).dropna()

        X = sm.add_constant(reg["lag"])

        model = sm.OLS(
            reg["y"],
            X
        ).fit()

        output.iloc[i] = float(
            model.params["lag"]
        )

    return output


# ------------------------------------------------------------
# CALCULATE ALL SIX
# ------------------------------------------------------------

rolling_phi = pd.DataFrame(
    index=spreads.index
)

for col in spreads.columns:

    rolling_phi[col] = (
        rolling_ar1_phi(
            spreads[col],
            WINDOW,
            MIN_OBS
        )
        .reindex(spreads.index)
    )


# ------------------------------------------------------------
# CURRENT PERSISTENCE TABLE
# ------------------------------------------------------------

rows = []

for col in spreads.columns:

    p = rolling_phi[col].dropna()

    if p.empty:
        continue

    phi = float(p.iloc[-1])
    date = p.index[-1]

    if 0 < phi < NEAR_UNIT_ROOT:

        half_life = (
            -np.log(2)
            / np.log(phi)
        )

        flag = "Stationary"

    elif NEAR_UNIT_ROOT <= phi < 1:

        half_life = np.nan
        flag = "Near unit root"

    elif phi >= 1:

        half_life = np.nan
        flag = "Non-stationary AR(1)"

    else:

        half_life = np.nan
        flag = "Negative AR coefficient"

    rows.append({
        "Spread": col,
        "As Of": date,
        "Rolling AR(1) φ": phi,
        "Half-Life (days)": half_life,
        "Persistence Flag": flag
    })


persistence = (
    pd.DataFrame(rows)
    .set_index("Spread")
)


print("=" * 90)
print("LATEST 252-OBSERVATION PERSISTENCE")
print("=" * 90)

display(
    persistence.style.format({
        "Rolling AR(1) φ": "{:.4f}",
        "Half-Life (days)": "{:.1f}"
    })
)


# ------------------------------------------------------------
# ONE PERSISTENCE CHART — PHI
# ------------------------------------------------------------

fig = go.Figure()

for col in rolling_phi.columns:

    s = rolling_phi[col].dropna()

    fig.add_trace(
        go.Scatter(
            x=s.index,
            y=s.values,
            mode="lines",
            name=col,
            line=dict(width=1.7)
        )
    )


# Unit-root reference
fig.add_hline(
    y=1.0,
    line_dash="dash",
    annotation_text="Unit root: φ = 1"
)

# Half-life reporting threshold
fig.add_hline(
    y=NEAR_UNIT_ROOT,
    line_dash="dot",
    annotation_text="Half-life threshold: φ = 0.995"
)


fig.update_layout(
    title="SA Rates Spreads — Rolling 252-Observation AR(1) Persistence",
    xaxis_title="Date",
    yaxis_title="AR(1) φ",
    hovermode="x unified",
    template="plotly_white",
    height=600,
    legend=dict(
        orientation="h",
        y=1.14
    )
)

fig.show()
```

```python
# ============================================================
# SA RATES RV MONITOR — FINAL COLAB OUTPUT
#
# Requires existing:
#   rates
#   spreads
#   rolling_phi
#
# Produces:
#   1. Final summary table
#   2. Spread history
#   3. Rolling 252D z-scores
#   4. Rolling AR(1) persistence
#   5. Final CSV outputs
# ============================================================

import numpy as np
import pandas as pd
import plotly.graph_objects as go
from IPython.display import display


# ============================================================
# 1. BUILD FINAL SUMMARY
# ============================================================

rows = []

for name in spreads.columns:

    s = spreads[name].dropna()

    current = float(s.iloc[-1])
    as_of = s.index[-1]

    # 60-observation change
    change_60 = (
        current - float(s.iloc[-61])
        if len(s) >= 61
        else np.nan
    )

    # 252-observation z-score
    trailing = s.tail(252)

    if len(trailing) >= 60 and trailing.std(ddof=1) > 0:
        z = (
            current - trailing.mean()
        ) / trailing.std(ddof=1)
    else:
        z = np.nan

    # Full-history percentile
    percentile = (
        (s <= current).sum()
        / len(s)
        * 100
    )

    # Latest rolling 252-observation AR(1)
    rp = rolling_phi[name].dropna()

    if not rp.empty:

        phi = float(rp.iloc[-1])
        phi_date = rp.index[-1]

    else:

        phi = np.nan
        phi_date = pd.NaT

    # Half-life / persistence classification
    if 0 < phi < 0.995:

        half_life = (
            -np.log(2)
            / np.log(phi)
        )

        persistence_flag = "Stationary"

    elif 0.995 <= phi < 1:

        half_life = np.nan
        persistence_flag = "Near unit root"

    elif phi >= 1:

        half_life = np.nan
        persistence_flag = "Non-stationary AR(1)"

    else:

        half_life = np.nan
        persistence_flag = "Negative / unavailable"

    rows.append({

        "Spread": name,

        "As Of": as_of.date(),

        "Current (bp)": current,

        "60D Δ (bp)": change_60,

        "252D Z": z,

        "Historical %ile": percentile,

        "Rolling AR(1) φ": phi,

        "Half-Life (days)": half_life,

        "Persistence": persistence_flag,

        "N": len(s)
    })


final_summary = (
    pd.DataFrame(rows)
    .set_index("Spread")
)


print("=" * 115)
print("SOUTH AFRICAN RATES RELATIVE-VALUE MONITOR")
print("=" * 115)

display(
    final_summary.style.format({

        "Current (bp)": "{:.2f}",

        "60D Δ (bp)": "{:+.2f}",

        "252D Z": "{:+.2f}",

        "Historical %ile": "{:.1f}",

        "Rolling AR(1) φ": "{:.4f}",

        "Half-Life (days)": "{:.1f}",

        "N": "{:.0f}"
    })
)


# ============================================================
# 2. CALCULATE ROLLING Z-SCORES
# ============================================================

rolling_mean = spreads.rolling(
    252,
    min_periods=60
).mean()

rolling_std = spreads.rolling(
    252,
    min_periods=60
).std()

rolling_z = (
    spreads - rolling_mean
) / rolling_std


# ============================================================
# 3. CHART — SAGB vs ZARONIA
# ============================================================

fig = go.Figure()

for col in [
    "1Y - ZARONIA",
    "5Y - ZARONIA",
    "10Y - ZARONIA"
]:

    s = spreads[col].dropna()

    fig.add_trace(
        go.Scatter(
            x=s.index,
            y=s.values,
            mode="lines",
            name=col,
            line=dict(width=1.8)
        )
    )

fig.add_hline(
    y=0,
    line_dash="dash"
)

fig.update_layout(
    title="SAGB Constant-Maturity Spreads vs ZARONIA",
    xaxis_title="Date",
    yaxis_title="Basis Points",
    hovermode="x unified",
    template="plotly_white",
    height=560,
    legend=dict(
        orientation="h",
        y=1.10
    )
)

fig.show()


# ============================================================
# 4. CHART — SAGB CURVE
# ============================================================

fig = go.Figure()

for col in [
    "5Y - 1Y",
    "10Y - 1Y",
    "10Y - 5Y"
]:

    s = spreads[col].dropna()

    fig.add_trace(
        go.Scatter(
            x=s.index,
            y=s.values,
            mode="lines",
            name=col,
            line=dict(width=1.8)
        )
    )

fig.add_hline(
    y=0,
    line_dash="dash"
)

fig.update_layout(
    title="SAGB Constant-Maturity Curve Spreads",
    xaxis_title="Date",
    yaxis_title="Basis Points",
    hovermode="x unified",
    template="plotly_white",
    height=560,
    legend=dict(
        orientation="h",
        y=1.10
    )
)

fig.show()


# ============================================================
# 5. CHART — ROLLING Z-SCORES
# ============================================================

fig = go.Figure()

for col in rolling_z.columns:

    s = rolling_z[col].dropna()

    fig.add_trace(
        go.Scatter(
            x=s.index,
            y=s.values,
            mode="lines",
            name=col,
            line=dict(width=1.5)
        )
    )

fig.add_hline(
    y=0,
    line_dash="dash"
)

fig.add_hline(
    y=2,
    line_dash="dot"
)

fig.add_hline(
    y=-2,
    line_dash="dot"
)

fig.update_layout(
    title="SA Rates RV — Rolling 252-Observation Z-Scores",
    xaxis_title="Date",
    yaxis_title="Z-Score",
    hovermode="x unified",
    template="plotly_white",
    height=600,
    legend=dict(
        orientation="h",
        y=1.12
    )
)

fig.show()


# ============================================================
# 6. CHART — ROLLING AR(1) PERSISTENCE
# ============================================================

fig = go.Figure()

for col in rolling_phi.columns:

    s = rolling_phi[col].dropna()

    fig.add_trace(
        go.Scatter(
            x=s.index,
            y=s.values,
            mode="lines",
            name=col,
            line=dict(width=1.7)
        )
    )

fig.add_hline(
    y=1.0,
    line_dash="dash",
    annotation_text="Unit root: φ = 1"
)

fig.add_hline(
    y=0.995,
    line_dash="dot",
    annotation_text="Half-life threshold: φ = 0.995"
)

fig.update_layout(
    title="SA Rates Spreads — Rolling 252-Observation AR(1) Persistence",
    xaxis_title="Date",
    yaxis_title="AR(1) φ",
    hovermode="x unified",
    template="plotly_white",
    height=600,
    legend=dict(
        orientation="h",
        y=1.12
    )
)

fig.show()


# ============================================================
# 7. SAVE FINAL OUTPUT
# ============================================================

dashboard_data = pd.concat(
    [
        rates.add_suffix("_RATE"),
        spreads.add_suffix("_SPREAD_BP"),
        rolling_z.add_suffix("_ZSCORE"),
        rolling_phi.add_suffix("_AR1_PHI")
    ],
    axis=1
)

dashboard_data.to_csv(
    "SA_Rates_RV_Final_Dashboard.csv"
)

final_summary.to_csv(
    "SA_Rates_RV_Final_Summary.csv"
)


print("\nSaved:")
print("SA_Rates_RV_Final_Dashboard.csv")
print("SA_Rates_RV_Final_Summary.csv")
```

```python
# ============================================================
# EXTRACT 1Y / 5Y / 10Y ZARONIA OIS BENCHMARK RATES
# ============================================================

import pandas as pd
import numpy as np

FILE = "/content/Benchmark_Rates.xlsb"

# Read using row 1 as headers
ois_raw = pd.read_excel(
    FILE,
    sheet_name="Benchmark Rates",
    engine="pyxlsb",
    header=1
)

print("Shape:", ois_raw.shape)

# ------------------------------------------------------------
# Identify exact maturity columns
# ------------------------------------------------------------

targets = ["Dates", "ZARONIA"]

for maturity in ["12M", "60M", "120M"]:
    matches = [
        c for c in ois_raw.columns
        if str(c).strip().endswith(f":{maturity}")
    ]

    print(maturity, "matches:", matches)

    if len(matches) != 1:
        raise ValueError(
            f"Expected exactly one {maturity} column, found {matches}"
        )

    targets.append(matches[0])


# ------------------------------------------------------------
# Extract
# ------------------------------------------------------------

ois = ois_raw[targets].copy()

ois.columns = [
    "Date",
    "ZARONIA",
    "OIS_1Y",
    "OIS_5Y",
    "OIS_10Y"
]


# Excel serial date -> datetime
ois["Date"] = pd.to_datetime(
    ois["Date"],
    unit="D",
    origin="1899-12-30"
)


# Decimal -> percentage rates
for c in [
    "ZARONIA",
    "OIS_1Y",
    "OIS_5Y",
    "OIS_10Y"
]:
    ois[c] = pd.to_numeric(
        ois[c],
        errors="coerce"
    ) * 100


ois = (
    ois
    .dropna(subset=["Date"])
    .set_index("Date")
    .sort_index()
)


# ------------------------------------------------------------
# VALIDATION OUTPUT
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("ZARONIA OIS TERM STRUCTURE")
print("=" * 80)

print("\nDate range:")
print(ois.index.min(), "to", ois.index.max())

print("\nObservations:")
print(ois.count())

print("\nFirst 5:")
display(ois.head())

print("\nLatest 10:")
display(ois.tail(10))

print("\nDescriptive statistics:")
display(ois.describe().round(3))
```
